```python
import sympy as sp

# Variables
m1 = sp.symbols('m_1') # masa m1 en kg
m2 = 3*m1
g = sp.symbols('g') # aceleración de la gravedad m/s^2

# Analisis de fuerzas
## Cuerpo 1
print('Fuerzas de cuerpo 1')
P1 = m1*g
T = sp.symbols('T')
a_1 = sp.symbols('a_1')
eq_1 = sp.Eq(m1*a_1, T - P1)
display(eq_1)
## Cuerpo 2
print('Fuerzas de cuerpo 2')
P2 = m2*g
a_2 = sp.symbols('a_2')
eq_2 = sp.Eq(m2*a_2, T - P2)
display(eq_2)
## Relación entre aceleraciones
print('Relación entre aceleraciones')
eq_3 = sp.Eq(- a_1, a_2)
display(eq_3)

# Solución de ecuaciones
print('Solución de ecuaciones')
sol = sp.solve([eq_1, eq_2, eq_3], (T, a_1, a_2))
display(sp.Eq(a_1, sol[a_1]))
display(sp.Eq(a_2, sol[a_2]))

# Funcion de posicion en función del tiempo
print('Función de posición en función del tiempo')
t = sp.symbols('t') # tiempo
x_1 = (a_1*t**2/2).subs(sol)
x_2 = (a_2*t**2/2).subs(sol)
display(sp.Eq(sp.Function('x_1')(t), x_1))
display(sp.Eq(sp.Function('x_2')(t), x_2))

# Tiempo de distancia entre cuerpos
print('Tiempo de distancia entre cuerpos tq x_1 - x_2 = h')
h = 0.5 # distancia entre cuerpos en metros
t_sol = sp.solve(sp.Eq(x_1 - x_2, h), t)
display(sp.Eq(sp.symbols('t_sol'), t_sol[1]))

# Velocidad de m1 en t_sol
print('Velocidad de m1 en t_sol')
v_1 = a_1*t_sol[1]
display(sp.Eq(sp.Function('v_1')(t), v_1))
v_1_num = v_1.subs({a_1: sol[a_1], t: t_sol[1], g: 9.81})
display(sp.Eq(sp.Function('v_1')(sp.symbols('t_sol')).subs(t, t_sol[1]), v_1_num).evalf(3))



```
