```python
import sympy as sp

# Datos
m = 90 # kg, masa del cuerpo
h = 50 # m, caida libre
v_l = 5 # m/s, velocidad limite

# Variables
W_f = sp.symbols('W_f') # Trabajo realizado por la fuerza de fricción
delta_k, delta_u = sp.symbols('Delta_K Delta_U') # Variación de energía cinética y potencial

print('El trabajo realizado por una fuerza no conservativa es igual a la variación de energía cinética y potencial')
eq_1 = sp.Eq(W_f, delta_k + delta_u)
display(eq_1)
eq_1 = eq_1.subs(delta_k, sp.symbols('m') * sp.symbols('v_f')**2 / 2 - sp.symbols('m') * sp.symbols('v_i')**2 / 2)
display(eq_1)
eq_1 = eq_1.subs(delta_u, sp.symbols('m') * sp.symbols('g') * sp.symbols('h_f') - sp.symbols('m') * sp.symbols('g') * sp.symbols('h_i'))
display(eq_1)
print('Donde:')
display(sp.Eq(sp.symbols('v_i'), 0))
display(sp.Eq(sp.symbols('v_f'), v_l))
display(sp.Eq(sp.symbols('h_i'), h))
display(sp.Eq(sp.symbols('h_f'), 0))

print('Despejando el trabajo realizado por la fuerza de fricción')
W_f = sp.solve(eq_1, W_f)[0]
display(
    W_f.subs(
        {
            sp.symbols('m'): m, 
            sp.symbols('g'): 9.81, 
            sp.symbols('v_i'): 0, 
            sp.symbols('v_f'): v_l, 
            sp.symbols('h_i'): h, 
            sp.symbols('h_f'): 0
        }
    ).evalf(2)
)
```
