```python
import sympy as sp

# Datos
a_0 = 2 # m/s^2 durante 25m, luego v constante

# Variables
t = sp.symbols('t')

# Funcion de posicion los primeros 25m
print('Funcion de posicion los primeros 25m corredor 1')
x_t = a_0*t**2/2
display(sp.Eq(sp.Symbol('x_{1}(t)'), x_t))
t_25m = sp.solve(x_t - 25, t)[1]
print('Funcion de velocidad los primeros 25m corredor 1')
v_t = sp.diff(x_t, t)
display(sp.Eq(sp.Symbol('v_{1}(t)'), v_t))
print('Velocidad en t_25m')
v_25m = v_t.subs(t, t_25m)
display(sp.Eq(sp.Symbol('v_{1}(t_{25m})'), v_25m))
print('funcion de posicion despues de los 25m')
x_t = 25 + v_25m*(t - t_25m)
display(sp.Eq(sp.Symbol('x_{1}(t)'), x_t))

# Funcion de posicion del corredor 2
print('Funcion de posicion corredor 2')
x_t = 8*t + 30
display(sp.Eq(sp.Symbol('x_{2}(t)'), x_t))

# Tiempo de encuentro
print('Tiempo de encuentro')
t_encuentro = sp.solve(sp.Eq(25 + v_25m*(t - t_25m), 8*t + 30), t)[0]
display(sp.Eq(sp.Symbol('t_{encuentro}'), t_encuentro))

# Posicion de encuentro
print('Posicion de encuentro')
x_encuentro = 25 + v_25m*(t_encuentro - t_25m)
display(sp.Eq(sp.Symbol('x_{encuentro}'), x_encuentro))

```
