# Ejercicio 1. Vector aceleracion en el movimiento circular.

Una partícula recorre la trayectoria que indica la línea de puntos de la figura.

![](ejercicio_0301.png)

Cuando se encuentra en el punto A, su rapidez esta disminuyendo.

Se pide:

- (I) Cual de los vectores que aparecen en ese punto representa mejor la aceleracion asociada al cambio en el modulo de la velocidad.

- (II) Cual representa mejor la aceleracion total a la cual esta sometida la particula considerando su trayectoria.

Lo primero a tener en cuenta es que en el punto A no estamos en un movimiento circular uniforme, ya que la rapidez de la partícula está disminuyendo. Por lo tanto, la aceleración total no es centrípeta.

La velocidad sin embargo sigue siendo tangente a la trayectoria.

El punto A incorpora una aceleracion nueva, que es responsable de la disminución de la rapidez. Esta aceleración tambien es tangente a la trayectoria, su direccion depende de si aumenta o disminuye la rapidez, su nombre es aceleracion tangencial.

La aceleracion centripeta no desaparece, sigue estando presente, pero no es la unica aceleracion presente en el punto A.

De esta forma tenemos:

- el vector velocidad $\vec{v}$ esta representado por el (b)
- el vector aceleracion tangencial $\vec{a}_t$ esta representado por el (e), dado que en el punto A la rapidez esta disminuyendo, el vector aceleracion tangencial esta en sentido contrario al vector velocidad y tangente a la trayectoria.
- el vector aceleracion centripeta $\vec{a}_c$ esta representado por el (c)
- el vector aceleracion total $\vec{a}$ esta representado por el (d), es la suma de los vectores (e) y (c)
