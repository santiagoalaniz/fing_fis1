# Ejercicio 6. Arrastre del agua.

Juan, que puede remar a 3.0 km/h respecto del agua, dirige su bote directamente a través de un río de 1.0 km de ancho, que corre a 2.0 km/h.

Se pide:

- A que distancia aguas abajo de su punto de partida, llega el bote a alcanzar la orilla opuesta? Que distancia ha recorrido con respecto a la orilla? Cuanto tiempo tarda en cruzar?

- Hacia donde debe dirigir su bote si Juan desea llegar a la orilla opuesta justo enfrente de su punto de partida? Cuanto tiempo tarda en cruzar?

Recordar la ley de velocidades relativas:

$
    v_{ps} = v_{ps'} + v_{s's}
$

### Parte 1

```python
import sympy as sp
import numpy as np

# Constants
mod_v_boat = 3 # km/h
mod_v_water = 2 # km/h
river_width = 1 # km

# Symbols
i, j = sp.symbols('i j') # Unit vectors
t = sp.symbols('t') # Time

## boat's velocity
v_boat_i = mod_v_boat
v_boat_j = mod_v_water

## boat's position
r_boat_i = v_boat_i*t
r_boat_j = v_boat_j*t

## time to cross the river (x = 1 km)
t_cross = river_width/v_boat_i
x, y = r_boat_i.subs(t, t_cross), r_boat_j.subs(t, t_cross)
vector = sp.nsimplify(x*i+j*y, tolerance=1e-6)
print(f"Time to cross the river: {t_cross*60:.2f} min")
print(f"Boat's position at crossing:")
display(vector)
vector_mod = sp.sqrt(x**2 + y**2)
print(f"Boat's position at crossing (modulus): {vector_mod:.2f} km")
```

### Parte 2

```python
import math
# Datos proporcionados
v_r = 3.0  # velocidad de Juan respecto al agua en km/h
v_c = 2.0  # velocidad de la corriente del río en km/h
d = 1.0    # ancho del río en km

# Velocidad perpendicular necesaria para cruzar directamente al otro lado
v_perpendicular = math.sqrt(v_r**2 - v_c**2)

# Tiempo para cruzar el río con la velocidad perpendicular efectiva
t_perpendicular = d / v_perpendicular

# Ángulo necesario para compensar la corriente
theta = math.atan(v_c / v_perpendicular) * (180 / math.pi)  # Convertido a grados para mejor comprensión

print(f"Velocidad perpendicular necesaria: {v_perpendicular:.2f} km/h")
print(f"Tiempo para cruzar el río con velocidad perpendicular: {t_perpendicular * 60:.2f} minutos")
print(f"Ángulo necesario para compensar la corriente: {theta:.2f} grados")
```
