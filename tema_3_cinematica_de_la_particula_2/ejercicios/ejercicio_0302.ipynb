(I) Una partícula efectua un movimiento circular. Usando analisis dimensional, encuentra una relacion entre su aceleracion radial, su velocidad tangencial y el radio de su trayectoria. (II) La rueda gigante de Londres tiene un radio de 60m, cuando esta girando con velocidad angular constante completa una vuelta en 30 minutos para que los pasajeros puedan disfrutar de la vista. Encuentra la aceleracion radial de un pasajero en la rueda gigante.

- (a) Cual es la  (modulo y sentido) de un pasajero en el punto mas alto?
- (b) Cual es la  (modulo y sentido) de un pasajero en el punto mas bajo?

## Parte (I)

Repasando, el analisis dimencional, es una tecnica que se usa para encontrar relaciones entre las variables de un problema, mas que nada, en base a sus dimensiones fisicas.

Tenemos que asumir que la aceleracion radial $a$ se puede expresar en funcion de la velocidad tangencial $v$, el radio de la trayectoria $r$, y una constante $k$ adimensional.

$
    [a] = [f(k, v, r)] = [k v^m r^n]$ = $[v^m r^n] \\
    [a] = [v^m] [r^n] = [LT^{-1}]^m [L]^n = L^{m+n} T^{-m} \\
    [a] = L^{m+n} T^{-m} = L T^{-2}
$

Tenemos el siguiente sistema de ecuaciones:

$
    \begin{cases}
        m + n = 1 \\
        m = 2 \\
    \end{cases} \\
$

De donde obtenemos que $m = 2$ y $n = -1$. Por lo tanto, la aceleracion radial se puede expresar como:

$
    a = k \frac{v^2}{r}
$

Faltaria encontrar el valor de $k$, del teorico sabemos que $a = \frac{v^2}{r}$, por lo tanto, $k = 1$.

Si la velocidad se duplica, entonces $v' = 2v$, y la aceleracion radial se puede expresar como:

$
    a' = \frac{v'^2}{r} = \frac{(2v)^2}{r} = 4 \frac{v^2}{r} = 4a
$

Es decir, la aceleracion radial se cuadriplica.


## Parte (II)

```python
import numpy as np

# datos del ejercicio

radio_noria = 60        # metros
t_vuelta_completa = 30  # minutos

vel_angular = 2 * np.pi / (t_vuelta_completa * 60)  # radianes por segundo
mod_vel_tan = radio_noria * vel_angular

mod_acc_cent = (mod_vel_tan ** 2) / radio_noria

print(f"Velocidad angular: {vel_angular:.4f} rad/s")
print(f"Módulo de la velocidad tangencial: {mod_vel_tan:.6f} m/s")
print(f"Módulo de la aceleración centrípeta: {mod_acc_cent:.6f} m/s^2")
print(f"Sentido de la aceleracion en el punto más alto: hacia el centro de la noria (abajo)")
print(f"Sentido de la aceleracion en el punto más bajo: hacia el centro de la noria (arriba)")
```
