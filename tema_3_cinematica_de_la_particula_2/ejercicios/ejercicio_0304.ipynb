# Ejercicio 4, viajando mas comodamente en un tren rapido

El tren *Grand Vitesse* recorre la campiña francesa a una velocidad de $216$ km/h. El principio de inercia incomoda a los pasajeros:

Al frenar, los pasajeros se desplazan hacia adelante

Al acelerar, los pasajeros se desplazan hacia atrás


Si tomaran una curva, los pasajeros se desplazarian hacia el lado opuesto de la curva, esto es porque la aceleración centrípeta es hacia el centro de la curva.

Para evitar que los pasajeros padezcan el ultimo efecto, se ha calculado que la aceleración centrípeta máxima que puede soportar un pasajero es de $0.5g$, donde $g$ es la aceleración de la gravedad.

Se pide:

1. Calcular el radio mínimo de una curva que puede tomar el tren a la velocidad máxima, de manera que los pasajeros no sientan incomodidad.

```python
import sympy as sp

# Constantes
g = 9.81 # m/s^2 aceleración de la gravedad
vel = 216 # km/h velocidad del tren

# Aceleracion centripeta
print('Formula aceleración centripeta')
a = sp.Symbol('v')**2 / sp.Symbol('r')
display(a)

# Aceleracion maxima permitida
print('Aceleración máxima permitida')
a_max = sp.Symbol('a_max')
a_max_eq = sp.Eq(0.05*sp.Symbol('g'), a_max)
display(a_max_eq)
a_max = 0.05 * g

# Radio de la curva
print('Formula radio de la curva')
r_eq = sp.Eq(sp.Symbol('r'), sp.Symbol('v')**2 / (sp.Symbol('g') * sp.Symbol('a_max')))
display(r_eq)

# Valor del radio
r = (vel * 1000 / 3600)**2 / (a_max)
print(f'Radio de la curva: {r/1000:.2f} km')



```
