# Ejercicio 5. Suma y resta de vectores, modulo de la velocidad/rapidéz.

En un rio la corriente va hacia el noroeste, a 10 km/h. Se observa un bote que se mueve hacia el sur a 20 km/h con respecto a la orilla.

Se pide:

- Calcular la velocidad/rapidez del bote con respecto al agua.

## Solucion

Hay dos sistemas de referencia, el rio y la orilla ($s$ y $s'$), ambos en velocidad constante (estar en reposo es un caso particular de velocidad constante).

Hay una particula, el bote ($p$), que se mueve.

Podemos aplicar la ley de transformacion de velocidades:

$
    v_{ps} = v_{ps'} + v_{s's}
$

Donde:

- $v_{ps}$ es la velocidad del bote con respecto al rio.
- $v_{ps'}$ es la velocidad del bote con respecto a la orilla.
- $v_{s's}$ es la velocidad de la corriente.

```python
import sympy as sp
import numpy as np

# Constantes
m_vp = 20 # km/h barco
m_vs = 10 # km/h corriente

# Variables
i, j = sp.symbols('i j') # Sistema N-E  

# Vectores de velocidades relativas
# desde el punto de vista de la orilla, el barco se mueve hacia el sur
vps_ =  -m_vp * j
# desde el punto de vista del agua, la orilla se mueve en la direccion contraria a la corriente
vs_s = (-m_vs * np.cos(np.pi/4) * i - m_vs * np.sin(np.pi/4) * j).evalf(4)
vps = vps_ + vs_s # solucion
coef_vps = [vps.coeff(v) for v in (i, j)]
m_vps = sp.sqrt(sum(c**2 for c in coef_vps))

# Simbolos
_vps_ = sp.symbols('v_{ps\'}')
_vs_s = sp.symbols('v_{s\'s}')
_vps = sp.symbols('v_{ps}')

# Ecuaciones
display(sp.Eq(_vps, _vps_ + _vs_s))
display(sp.Eq(_vps_, vps_))
display(sp.Eq(_vs_s, vs_s))

# Solucion
print('Velocidad relativa del barco respecto a la orilla:')
display(sp.Eq(_vps, vps))
display(sp.Eq(sp.symbols('|v_{ps}|'), m_vps))


```
