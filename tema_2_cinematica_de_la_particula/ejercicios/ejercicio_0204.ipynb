# Ejercicio 4. Caida libre con retardo temporal, trabajo con parametros.

Se dejan caer dos esferas pesadas, desde distintas alturas, la segunda $t_0$ segundos después de la primera.

Si las dos llegan al suelo al mismo tiempo $t_f$ despues de haber soltado la primera.

¿Desde que altura se soltó la segunda esfera?

```python
import sympy as sp

# time variable
t = sp.symbols('t')

# Initial position sphere_a, sphere_b
r0_a = sp.symbols('r0_a')
r0_b = sp.symbols('r0_b')

# Constant acceleration sphere_a, sphere_b
a = sp.symbols('g')

# Position funcions for sphere_a
ra_t = r0_a - (1/2)*a*t**2 # v0_a = 0, because it is dropped

# sphere_b is dropped t_0 seconds after sphere_a
t_0 = sp.symbols('t_0')

# Position funcions for sphere_b, t >= t_0
rb_t = r0_b - (1/2)*a*(t - t_0)**2 # v0_b = 0, because it is dropped

# time when both spheres hit the ground
t_f = sp.symbols('t_f')

# solve r0_a in function of t_f
r0_a = sp.solve(ra_t.subs(t, t_f), r0_a)[0]
display(sp.simplify(r0_a))

# solve r0_b in function of t_f
r0_b = sp.solve(rb_t.subs(t, t_f), r0_b)[0]
display(sp.simplify(r0_b))
```
