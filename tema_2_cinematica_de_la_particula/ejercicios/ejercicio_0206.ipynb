# Ejercicio 6. Choque en caida libre con retardo temporal.

Se tira una piedra desde el piso verticalmente y hacia arriba con velocidad 16 m/s. Dos segundos después se tira otra piedra, desde el piso.

Se pide:

- Calcular la velocidad con la que debe ser tirada para que la segunda piedra choque con la primera a una altura de 9 m.

- Calcular el tiempo que transcurre desde que se tiro la primera piedra hasta el choque.

```python
import sympy as sp

# Time variable
t = sp.symbols('t')

# Constants
G = 9.81 # m/s^2

# Initial conditions

## Initial Positions
r_a0 = 0 # from the ground
r_b0 = 0 # from the ground

## Initial Velocities
v_a0 = 16 # given
v_b0 = sp.symbols('v_b0') # unknown

## Initial Accelerations
a_a0 = -G # gravity
a_b0 = -G # gravity

# Functions
r_a = r_a0 + v_a0*t + 0.5*a_a0*t**2
r_b = r_b0 + v_b0*(t-2) + 0.5*a_b0*(t-2)**2 # 2sec delay

# Solve when r_a = 9m
t_sol = sp.solve(r_a-9,t)

# get t_sol[1] because t_sol[0] is less than 2 sec
t_sol = t_sol[1]

print(f"t = {t_sol} sec")

# Solve for v_b0
v_b0 = sp.solve(r_b.subs(t,t_sol)-9,v_b0)

print(f"v_b0 = {v_b0[0]} m/s")
```
