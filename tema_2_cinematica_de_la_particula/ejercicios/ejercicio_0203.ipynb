# Ejercicio 3. Calculo de la velocidad y posicion.

Una particula sale del origen en $t=0$ con una velocidad inicial de $\vec{v_0} = 3.6i$ y experimenta una aceleracion de $\vec{a}(t) = -1.2i - 1.4tj$.

Se pide:

- (a) Escribe las expresiones de los vectores $\vec{v}(t)$ y $\vec{r}(t)$.
- (b) En que tiempo llega la particula a su coordenada $x$ maxima.
- (c) Cual es la velocidad de la particula en ese momento.
- (d) Donde esta la particula en ese momento.

## Solución

Vamos a utilizar `sympy`, una libreria de python que nos permite hacer calculo simbolico.

Es util para definir las variables atomicas del problema.

- $t$: tiempo
- $i, j$: vectores unitarios

Notar que en este ejercicio la coordenada $k$ no aparece, por lo que no es necesario definirla.

```python
import sympy as sp

# Time variable
t = sp.symbols('t')

# Coordinates
i, j  = sp.symbols('i j')
```

### parte (a)

Integramos la aceleracion tomando en cuenta los valores iniciales de la velocidad y la posicion.

```python
# Initial velocity
v_0 = 3.6*i

# Acceleration vector
a_t = -1.2*i - 1.4*t*j

# Velocity vector
v_t = sp.integrate(a_t, t) + v_0

# Position vector 
r_t = sp.integrate(v_t, t) # + r_0 (0, 0, 0)

display(v_t, r_t)
```

### parte (b)

Encontrar el tiempo en que la particula llega a su coordenada $x$ maxima, es equivalente a encontrar el tiempo en que la velocidad en $x$ es cero.

Como estamos trabajando con el sistema de coordenadas $i, j$, la funcion de velocidad en $x$ es $v_i(t)$.

$t_{x_{max}} = v_i(t) = 0 \rightarrow -1.2*t + 3.6 = 0$

```python
t_x_max = sp.solve(-1.2*t + 3.6, t)

display(t_x_max[0])
```

### parte (c)

$t_{x_{max}} = 3s$, entonces la velocidad de la particula en ese momento es $\vec{v}(3s)$


```python
v_x_max = v_t.subs(t, t_x_max[0])

v_x_max_nsimplified = sp.nsimplify(v_x_max, tolerance=1e-6)
display(v_x_max, v_x_max_nsimplified)
```

### parte (d)

La posicion de la particula en ese momento es $\vec{r}(3s)$

```python
r_x_max = r_t.subs(t, t_x_max[0])

display(r_x_max)
```
