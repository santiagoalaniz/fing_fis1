# Ejercicio 1. Caida libre, relación entre variables.

Se arrojan directamente hacia los objetos A y B. La rapidez inicial del objeto A es cuatro veces mayor que la del objeto B. ¿Cuál de las siguientes afirmaciones es correcta?

- (a) la mitad del tiempo que tarda B. 
- (b) cuatro veces el tiempo que tarda B. 
- (c) la cuarta parte de lo que tarda B. 
- (d) el doble de lo que tarda B. 
- (e) igual que lo que tarda B. 

![](ejercicio_0201.png)

## Solucion

Los datos iniciales son:

$
\begin{aligned}
y_{A_0} &= y_{B_0} = 0 \\
v_{A_0} &= 4 v_{B_0} \\
a_{A}(t) &= a_{B}(t) = -g
\end{aligned}
$

La ecuación de la posición en función del tiempo para un objeto en caída libre es:

$
\begin{aligned}
y_A(t) &= y_{A_0} + 4 v_{B_0} t - \frac{1}{2} g t^2 \\
y_B(t) &= y_{B_0} + v_{B_0} t - \frac{1}{2} g t^2 \\
\end{aligned}
$

El tiempo final $t_f$ es un valor estríctamente positivo donde $y_A(t_{f_A}) = y_B(t_{f_B}) = 0$. Es decir, el tiempo que tarda en volver a llegar al suelo.

$
\begin{aligned}
y_A(t_{f_A}) &= 0 \rightarrow 4 v_{B_0} t_{f_A} - \frac{1}{2} g t_{f_A}^2 = 0 \\
y_B(t_{f_B}) &= 0 \rightarrow v_{B_0} t_{f_B} - \frac{1}{2} g t_{f_B}^2 = 0 \\
t_{f_A}, t_{f_B} &> 0
\end{aligned}
$

Despejando $t_{f_A}$ y $t_{f_B}$:

$
\begin{aligned}
t_{f_A} &= \frac{8 v_{B_0}}{g} \\
t_{f_B} &= \frac{2 v_{B_0}}{g} \\
\end{aligned}
$

Finalmente, la relación entre los tiempos es:

$
\begin{aligned}
\frac{t_{f_A}}{t_{f_B}} &= \frac{8 v_{B_0}}{g} \frac{g}{2 v_{B_0}} = 4 \\
t_{f_A} &= 4 t_{f_B}
\end{aligned}
$
