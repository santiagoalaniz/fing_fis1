# Problema 7: Cuidado con la Llave!

Un tecnico esta a 60 metros por encima de un asensor, este se enciende lo que hace al operador sobresaltarse y dejar caer una llave.

El acensor se mueva hacia arriba con una aceleracion de 0.1 g los primeros 3 segundos, luego se mueve con velocidad constante.

Se pide:

- Calcular la altura en la que la llave choca con el ascensor.

```python
import sympy as sp

# Definir símbolos
t = sp.symbols('t')
g = 9.8  # Aceleración debido a la gravedad

# Altura inicial de la llave
h0 = 60

# Ecuaciones de movimiento
h_llave = h0 - (1/2)*g*t**2

# Fase 1 del ascensor (0 a 3 segundos)
a1 = 0.1*g
h_ascensor1 = (1/2)*a1*t**2

# Velocidad del ascensor después de 3 segundos
v = a1*3

# Fase 2 del ascensor (después de 3 segundos)
h_ascensor1_at_3 = (1/2)*a1*3**2
h_ascensor2 = h_ascensor1_at_3 + v*(t - 3)

# Resolver para t en ambas fases
# Fase 1
t_encuentro_fase1 = sp.solve(-h_llave + h_ascensor1, t)

# Fase 2
t_encuentro_fase2 = sp.solve(-h_llave + h_ascensor2, t)

t_validos = [t_val for t_val in t_encuentro_fase1 if t_val.is_real and t_val > 0 and t_val < 3]

if t_validos:
    t_encuentro = t_validos[0]
    altura_encuentro = h_llave.subs(t, t_encuentro)
    print("El encuentro ocurre a los {:.2f} segundos a una altura de {:.2f} metros.".format(t_encuentro, altura_encuentro))
else:
    print("No se encontró un tiempo válido de encuentro en fase 1.")

# Verificar cuál tiempo es válido.
t_validos = [t_val for t_val in t_encuentro_fase2 if t_val.is_real and t_val > 3]

if t_validos:
    t_encuentro = t_validos[0]
    altura_encuentro = h_llave.subs(t, t_encuentro)
    print("El encuentro ocurre a los {:.2f} segundos a una altura de {:.2f} metros.".format(t_encuentro, altura_encuentro))
else:
    print("No se encontró un tiempo válido de encuentro. en fase 2")
```
