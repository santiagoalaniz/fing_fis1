# Ejercicio 2. Cálculo de velocidad y aceleración

Una particula se mueve de modo que su posición en el tiempo $t$ está dada por la función vectorial:

$\vec{r}(t) = i + 4t^2j + tk$

Se pide:

- (a) Escribir las expresiones para la velocidad.
- (b) Escribir las expresiones para la aceleración.
- (c) Cual es la forma de la trayectoria de la partícula.

```python
import sympy as sp
from sympy.vector import CoordSys3D

# t variable
t = sp.symbols('t')

# Unit vectors
uv = CoordSys3D('')
i, j, k = uv.i, uv.j, uv.k

# Position vector
r = i + 4*t**2*j + t*k
# Velocity vector
v = sp.diff(r, t)
# Acceleration vector
a = sp.diff(v, t)

display(r, v, a)
```

La forma de la trayectoria de la partícula es una parábola en el plano $yz$.

- $x = 1$
- $y = 4t^2$
- $z = t$

La ecuación de la parábola es:

$y(z) = 4z^2$

La partícula se mueve de la siguiente manera:

- $x$ es constante, siemre es 1.
- $y$ es una parábola en función de $z$.
- $z$ es una variable que crece linealmente con el tiempo.
