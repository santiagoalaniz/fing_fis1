# Ejercicio 5. El hombre bala.

En un circo, un Hombre Bala sale de un cañon y debe aterrizar en una red a $L = 10m$ bajo la boca del cañon. Si las componentes de velocidad inicial son:

- $v_{0x} = 10 m/s$ en la dirección horizontal.
- $v_{0y} = 20 m/s$ en la dirección vertical.

Se pide calcular:

- Cuanto tiempo esta en el aire.
- Donde debe estar la red.

Considera ahora que a una distancia horizontal de $\frac{3}{2}L$ de la boca del cañon hay un muro de $\frac{5}{2}L$ de altura.

Se pide calcular:

- Si el hombre bala choca contra el muro o lo sobrepasa.

```python
import sympy as sp

# Time
t = sp.symbols('t')

# Constants
g = 9.81 # m/s^2
L = 10

# Initial position
r_0_x, r_0_y = 0, L # m

# Initial velocity
v_0x, v_0y = 10, 20 # m/s

# r_vec(t) components
r_x, r_y = r_0_x + v_0x*t, r_0_y + v_0y*t - (g/2)*t**2
print('componetes de r(t): rx, ry')
display(r_x, r_y)

# flight time
t_flight = sp.solve(r_y, t)[1] # positive root
print('flight time (s):')
display(t_flight) 

# net position
r_x_net = r_x.subs(t, t_flight)
print('net position: (x, y)')
display(r_x_net)
display(0)

# time to reach the wall
t_wall = sp.solve(r_x - (3/2)*L, t)[0]
print('time to reach the wall (s):')
display(t_wall)

# height at the wall
r_y_wall = r_y.subs(t, t_wall)
print('height at the wall (m):')
display(r_y_wall)

# impact on the wall?
print(f"impact on the wall: {r_y_wall < (5/2 * L)}")
```
