# Problema 8, Movimiento de Proyectil.

Una bolita rueda sobre una mesa horizontal de altura $h$ con una rapidez $v_0$.

La bolita rebota elásticamente en una pared vertical a distancia $D$ de la orilla de la mesa.

Despues, la bolita llega al piso a una distancia $x_0$ de la orilla de la mesa.

Se pide:

- Hallar las ecuaciones para las ordenadas y las abscisas de la bolita en función del tiempo, validas para antes de chocar con la pared.
- Determina $s$ y el valor de $v_y$ cuando la bolita llega al muro.
- Halla una expresion para $x_0$ en funcion de $h$, $D$ y $v_0$ y determina el valor de v0 para que $x_0 = 0$. ¿Cual es el valor de $s$ en ese caso?

**Nota**:

*Elasticamente significa que la $v_y$ no cambia y la $v_x$ se invierte.*

```python
import sympy as sp

## Variables
t = sp.symbols('t')     # time
i,j = sp.symbols('i j') # unit vectors

## Constants
h = sp.symbols('h')     # height (j direction)
D = sp.symbols('D')     # wall distance from origin (i direction)
v_0 = sp.symbols('v_0') # initial velocity (i direction)
g = sp.symbols('g')     # gravity acceleration (j direction)

print('primeira parte')

## position before hitting the wall
r_i = v_0*t
r_j = h - 0.5*g*t**2

display(sp.nsimplify(r_i))
display(sp.nsimplify(r_j))

print('segunda parte')

### s_wall
s_wall = sp.solve(r_i - D, t)[0] # time when the ball hits the wall
display(s_wall)

### v_j_wall
v_j_wall = r_j.diff(t).subs(t, s_wall) 
display(sp.nsimplify(v_j_wall))

### r_j at s_wall
r_j_wall = r_j.subs(t, s_wall) # j-position when the ball hits the wall
display(sp.nsimplify(r_j_wall))

print('tercera parte')

r_j_aw = r_j_wall - 0.5*g*(t - s_wall)**2
display(sp.nsimplify(r_j_aw))

r_i_aw = D - v_0*(t - s_wall)
display(sp.nsimplify(r_i_aw))

r_j_ground = sp.solve(r_j_aw, t)[1] # time when the ball hits the ground
display(sp.nsimplify(r_j_ground))

s_ground = r_i_aw.subs(t, r_j_ground) # i-position when the ball hits the ground
display(sp.nsimplify(s_ground))

v_0_ground = sp.solve(s_ground, v_0)[0] # initial velocity for r_i_ground = 0
display(sp.nsimplify(v_0))


s_total = s_wall + s_ground

s_total = s_total.subs(v_0, v_0_ground)

display(sp.nsimplify(s_total))









```
