# Ejercicio 2, Energia Cinetica.

Un hombre y un niño corren. El hombre tiene la mitad de la energía cinética del niño. El hombre tiene el doble de masa que el niño. El hombre aumenta su velocidad en 1 m/s de modo que ahora tiene la misma energía cinética que el niño. ¿Cuáles eran las velocidades iniciales del hombre y del niño?

```python
import sympy as sp

# Variables
## Masa del niño
m_nino = sp.symbols("m_{niño}")
## Masa del hombre
m_hombre = 2 * m_nino
## Velocidad del niño
v_nino = sp.symbols("v_{niño}")
## Velocidad del hombre
v_hombre = sp.symbols("v_{hombre}")
## Energía cinética del niño
K_nino = sp.symbols("K_{niño}")
k_nino = sp.Rational(1, 2) * m_nino * v_nino**2
display(sp.Eq(K_nino, k_nino))
## Energía cinética del hombre (mitad de la del niño)
K_hombre = sp.symbols("K_{hombre}")
k_hombre = k_nino / 2
display(sp.Eq(K_hombre, k_hombre))
## Energia cinética del hombre en función de su velocidad
k_hombre_2 = sp.Rational(1, 2) * m_hombre * v_hombre**2
display(sp.Eq(K_hombre, k_hombre_2))

eq_v_hombre = sp.Eq(k_hombre_2, k_hombre)
display(eq_v_hombre)
v_hombre_sol = sp.solve(eq_v_hombre, v_hombre)[1]
display(sp.Eq(v_hombre, v_hombre_sol))

# El hombre aumenta su velocidad en 1 m/s
n_v_hombre_sol = v_hombre_sol + 1
display(sp.Eq(v_hombre, n_v_hombre_sol))
## Al hacer esto, ahora tiene la misma energía cinética que el niño
eq_v_nino = sp.Eq(k_nino, k_hombre_2.subs(v_hombre, n_v_hombre_sol))
display(eq_v_nino)
## Dividir la masa del niño de ambos lados
eq_v_nino = sp.Eq(eq_v_nino.lhs / m_nino, eq_v_nino.rhs / m_nino)
display(eq_v_nino)
v_nino_sol = sp.solve(eq_v_nino, v_nino)[1]
display(sp.Eq(v_nino, v_nino_sol))
v_hombre_sol = v_hombre_sol.subs(v_nino, v_nino_sol)
display(sp.Eq(v_hombre, v_hombre_sol))
```
