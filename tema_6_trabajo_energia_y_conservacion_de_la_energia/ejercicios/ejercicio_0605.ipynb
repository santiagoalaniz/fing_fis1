# Ejercicio 5, Potencia: trabajo por unidad de tiempo.

```python
import sympy as sp
import numpy as np

m = 1380    # kg
v = 1.34    # m/s
d_x = 39.4  # m
d_y = 28.2  # m
mu_k = 0.41 # coeficiente de fricción cinética

# Calcular el trabajo neto.
print("Trabajo neto: 0 (hay velocidad constante)")

# Trabajo realizado por distintas fuerzas.
W_m, W_p, W_f = sp.symbols("W_m W_p W_f") # Trabajo realizado por motor, peso y fricción
eq_wneto = sp.Eq(sp.Eq(W_m - (W_p + W_f), 0), sp.symbols("W_{neto}"))
display(eq_wneto)
d =  np.sqrt(d_x**2 + d_y**2) # m
theta = np.arctan(d_y/d_x) # rad
F_p, F_f = sp.symbols("F_p F_f")
eq_F_p = sp.Eq(F_p, m*9.81*np.sin(theta))
display(eq_F_p)
eq_F_f = sp.Eq(F_f, mu_k*m*9.81*np.cos(theta))
display(eq_F_f)

F_neg = F_p + F_f
eq_F_neg = sp.Eq(F_neg, eq_F_p.rhs + eq_F_f.rhs)
display(eq_F_neg)
W_fp = sp.symbols("W_{fp}")
eq_W_fp = sp.Eq(W_fp, -eq_F_neg.rhs*d)
display(eq_W_fp)
W_m = sp.symbols("W_m")
eq_W_m = sp.Eq(W_m, -eq_W_fp.rhs)
display(eq_W_m)
P_m = sp.symbols("P_m")
eq_P_m = sp.Eq(P_m, sp.sympify(eq_W_m.rhs/(d/v)).evalf(2))
display(eq_P_m)
```
