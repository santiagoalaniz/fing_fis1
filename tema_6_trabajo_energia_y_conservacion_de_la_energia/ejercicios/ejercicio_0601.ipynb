# Ejercicio 1, Trabajo de fuerzas que no dependen de la posición.

Para empujar una caja de 25 kg hacia arriba sobre un plano inclinado de 27 grados, un obrero ejerce una fuerza de 120 N, paralela al plano. Cuando la caja se ha deslizado 3.6 metros. ¿Cuánto trabajo se ha realizado sobre la caja?

- Por el obrero
- La gravedad
- La normal

```python
import sympy as sp
import sympy.vector as spv
import numpy as np

# Datos
## Masa de la caja
m = 25 # kg
## Grado de inclinación del plano
theta = 27 * np.pi/180 # rad
## Modulo de fuerza del obrero
mod_F = 120 # N
## Modulo de la distancia
mod_d = 3.6 # m
## Aceleracion de la gravedad
g = 9.81 # m/s^2

# Recordar: El trabajo es el producto escalar de la fuerza y el desplazamiento
print("Trabajo de una Fuerza")
eq_w = sp.Eq(sp.symbols('W'), sp.symbols('|F|')*sp.symbols('|d|')*sp.cos(sp.symbols('theta')))
display(eq_w)

# Variables
print('Eje de referencia, plano inclinado de 27 grados \n')
## Sistema de coordenadas
N = spv.CoordSys3D('')
i, j = N.i, N.j

# Representacion vectorial de las fuerzas, aunque no es necesario
## Fuerza aplicada por el obrero
print("Fuerza aplicada por el obrero")
F_vec = mod_F*i + 0*j
display(F_vec)

## Fuerza de la normal
print("Fuerza de Normal")
N_vec = m*g*j
display(N_vec)

## Fuerza del Peso
print("Fuerza del Peso, tomando en cuenta el plano inclinado")
P_vec = -m*g*sp.sin(theta)*i - m*g*sp.cos(theta)*j
display(P_vec)

# Representacion vectorial del desplazamiento, aunque no es necesario
print("Desplazamiento")
d_vec = mod_d*i
display(d_vec)

# Trabajo realizado por la fuerza aplicada por el obrero
print("Trabajo realizado por la fuerza aplicada por el obrero")
theta_F_d = 0 # el mismo angulo
W_F = 120 * 3.6 * sp.cos(theta_F_d)
display(W_F)

# Trabajo realizado por la fuerza de la normal
print("Trabajo realizado por la fuerza de la normal")
theta_N_d = np.pi/2  # 90 grados
W_N = 0
display(W_N)

# Trabajo realizado por la fuerza del peso
print("Trabajo realizado por la fuerza del peso")
theta_P_d = np.pi/2 + theta # 90 grados + 27 grados
W_P = m*g*mod_d*sp.cos(theta_P_d)
display(W_P)

```
