# Ejercicio 3, Teorema del trabajo y la energía cinética.

Un bloque de masa m se deja caer desde una altura H sobre un resorte vertical (sin masa) de constante elástica k. El bloque se adhiere al resorte y se comprime una distancia $\Delta_l$ antes de alcanzar el reposo. Se desprecia el rozamiento.

- a) Calcula el trabajo que efectua la fuerza de gravedad sobre la masa desde que entra en contacto con el resorte hasta que llega a su punto de maxima comprension. Cuanto trabajo realiza la fuerza elastica sobre la masa durante este mismo intervalo.

- b) ¿Cuál era la velocidad v0 del bloque inmediatamente antes de alcanzar el resorte, en función de Δl y otros datos mencionados en la letra?

- c) Si el bloque tiene una masa de 263g, la constante elasteica del resorte es de 2.52 N/cm y la comprension maxima del resorte es de 11.8 cm, desde que altura se dejo caer el bloque?

- d) Si el bloque se deja caer desde una altura de 4H, cuanto vale la compresion maxima del resorte?

```python
import sympy as sp

# Variables
## masa
m = sp.Symbol('m')  # kg
## altura inicial
h = sp.Symbol('h')  # m
## distancia de compresion
delta_l = sp.Symbol('Delta_l')  # m
## constante elastica
k = sp.Symbol('k') # N/m
## aceleracion de la gravedad
g = sp.Symbol('g') # m/s^2

# Direccion y sentido: j (N-S) osea hacia abajo es positivo, hacia arriba es negativo

# Parte A
display("Parte A")
# Trabajo realizado por la fuerza Peso cuando el resorte se comprime
# El peso es una fuerza conservativa, por lo que el trabajo es igual 
# a la diferencia de energía potencial gravitatoria.
display("Formula de la energia potencial gravitatoria")
eq_U_g = sp.Eq(sp.Symbol('U_g'), m*g*h)
display(eq_U_g)

display("Trabajo del Peso, la fuerza Peso coincide con la direccion, i.e signo positivo")
eq_W_P = sp.Eq(sp.Symbol('W_P'), m*g*(delta_l))
display(eq_W_P)

# Trabajo realizado por la fuerza elástica
# La fuerza elástica es una fuerza conservativa, por lo que el trabajo es igual
# a la diferencia de energía potencial elástica.
display("Formula de la energia potencial elastica")
eq_U_e = sp.Eq(sp.Symbol('U_e'), k*sp.Symbol('x')**2/2)
display(eq_U_e)

display("Trabajo de la fuerza elastica, la fuerza elastica se opone al desplazamiento i.e signo negativo")
eq_W_e = sp.Eq(sp.Symbol('W_e'), -k*(delta_l**2)/2)
display(eq_W_e)

# Parte B
display("Parte B")
# Calcular la velocidad del bloque antes de que este se comprima es equivalente a calcular la velocidad
# del bloque ya comprimido.



```
