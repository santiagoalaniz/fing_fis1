# Ejercicio 4, Sistema de poleas: menos fuerza, mayor distancia.

```python
import sympy as sp

# Formula de trabajo
W, F, d, theta = sp.symbols('W F d theta')
eq = sp.Eq(W, F * d * sp.cos(theta))
display(eq)

# a)
print('(a) Trabajo de F si no estuvieran las poleas')
h = sp.symbols('h') # Altura
PP = sp.symbols('P_P') # Peso de la polea
PL = sp.symbols('P_L') # Peso de la carga
theta_val = 0 # Angulo entre F y d
display(eq.subs({F: (PP+PL), d: h, theta: theta_val}))

# b)
print('(b) Fuerza minima para levantar la carga')
# Analisis de fuerzas
P = PP + PL
F = sp.symbols('F')
eq_af = sp.Eq(F-P, 0)
display(eq_af)
# Actuan cuatro Tensiones sobre el sistema carga-polea
T = sp.symbols('T')
eq_af = sp.Eq(sp.symbols('F_{min}'), eq_af).subs(F, 4*T)
display(eq_af)
eq_af = sp.Eq(sp.symbols('F_{min}'), (PP + PL)/4)
display(eq_af)

# c)
print('(c) A traves de que distancia debe aplicarse la fuerza F para levantar la carga una altura h')
# U_g cuando la carga esta a una altura h
U_g = sp.symbols('U_g')
eq_ug = sp.Eq(sp.Eq(sp.symbols('\Delta_{U_g}'), (PL+PP)*h - sp.symbols('0')*(PL+PP)), W) 
display(eq_ug)
# Teorema de trabajo y energia
eq_twe = sp.Eq(U_g, W)
display(eq_twe)
eq_twe = eq_twe.subs({U_g: (PL+PP)*h, W: F*d})
display(eq_twe)
eq_twe = eq_twe.subs({F: (PP+PL)/4})
display(eq_twe)
# Despejando d
eq_twe = sp.solve(eq_twe, d)
display(sp.Eq(d, eq_twe[0]))

# d)
print('(d) Trabajo de F si estuvieran las poleas')
d = eq_twe[0]
F = ((PL+PP)/4)
eq_w = sp.Eq(W, F*d)
display(eq_w)
display(sp.simplify(eq_w))







```
