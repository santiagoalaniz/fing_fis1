# Ejercicio 5: Radar, descomposicion vectorial

Un radar detecta un misil que se aproxima desde el este al oeste. En el primer contacto la distancia del radar al cohete es de 40km a 40 grados sobre el horizonte. El cohete es rastreado por el radar durante 123 grados en el plano formado por la linea este-oeste y la vertical siendo la distancia del contacto final de 80km. Hallar el desplazamiento del cohete durante el rastreo.

![](ejercicio_0105.png)

```python
import numpy as np

ALPHA = 40
THETA = ALPHA + 123
FIRST_READ = 40
SECOND_READ = 88

alpha_rad = np.deg2rad(ALPHA)
x1, x2 = FIRST_READ * np.cos(alpha_rad), FIRST_READ * np.sin(alpha_rad)

theta_rad = np.deg2rad(THETA)
y1, y2 = SECOND_READ * np.cos(theta_rad), SECOND_READ * np.sin(theta_rad)
d1, d2 = x1 - y1, x2 - y2
rd = np.sqrt(d1**2 + d2**2)

print(f'x1: {x1:.2f}, x2: {x2:.2f}')
print(f'y1: {y1:.2f}, y2: {y2:.2f}')
print(f'd1: {d1:.2f}, d2: {d2:.2f}')
print(f'rd: {rd:.2f}')
```
