# Ejercicio 7: Corner en el estadio de fútbol.

Considera  coordenadas  (x,y)  en  el  plano  horizontal  de  la  cancha  del  Estadio  Centenario.  El  origen de coordenadas coincide con el banderín de corner situado entre la tribuna Ámsterdam y Olímpica y las coordenadas (x,y) con la línea lateral y de meta, respectivamente. 

Un córner pateado desde el origen, provoca que la proyección del desplazamiento de la pelota sobre el césped  forme  un  ángulo  de  28,2º  con  el  lateral,  hasta  que  un  segundo  jugador  que  se encuentra a 20,7 m del banderín, la desvía un ángulo de 105º, con el objetivo de hacer un gol. 

Si las coordenadas de los palos del arco son 30,34 y 37,66, medidas en metros, ¿tiene alguna chance de convertir? Si tu respuesta es negativa, indica el intervalo de ángulos que debió haber sido desviada la pelota.

Nota: El desvío de 105º es respecto de la línea que seguiría la pelota si no hubiera sido desviada.

![](ejercicio_0107.png)

```python
import numpy as np

PRIMER_PALO = 0, 30.34
SEGUNDO_PALO = 0, 37.66

D_JUGADOR_CORNER = 20.7
A_PELOTA_CORNER = 28.8 / 180 * np.pi
A_DESVIO_JUGADOR = (105 / 180 * np.pi)
DESVIO_TOTAL = A_PELOTA_CORNER + A_DESVIO_JUGADOR

pos_jugador = np.cos(A_PELOTA_CORNER) * D_JUGADOR_CORNER, np.sin(A_PELOTA_CORNER) * D_JUGADOR_CORNER
m = np.tan(DESVIO_TOTAL)

def f(x): return m * x + pos_jugador[1] - m * pos_jugador[0]

print(f"--- Escenario de tiro de esquina planteado en la letra ---")
print(f"El arco esta sobre el eje y, en las cordenadas {PRIMER_PALO} y {SEGUNDO_PALO}")
print(f"Condicion de gol: cualquier punto en el segmento de recta que define {PRIMER_PALO} y {SEGUNDO_PALO}")
print(f"La pelota es pateada desde el corner, viene con un angulo de {A_PELOTA_CORNER * 180 / np.pi:.2f} grados")
print(f"La pelota la recibe el jugador en el punto ({pos_jugador[0]:.2f}, {pos_jugador[1]:.2f})")
print(f"El jugador desvia la pelota {A_DESVIO_JUGADOR * 180 / np.pi:.2f} grados")
print(f"La recta que describe el desvio total (jugador + corner) es y = {m:.2f}x + {pos_jugador[1] - m * pos_jugador[0]:.2f}")
print(f"La pelota pasa por la linea de meta en el punto (0, {f(0):.2f})")
print(f"--- Conclusion: El jugador fallara si el desvio es el planteado ---")

m_primer_palo = (PRIMER_PALO[1] - pos_jugador[1]) / (pos_jugador[0] - PRIMER_PALO[0])
a_primer_palo = np.arctan(m_primer_palo) * 180 / np.pi
m_segundo_palo = (SEGUNDO_PALO[1] - pos_jugador[1]) / (pos_jugador[0] - SEGUNDO_PALO[0])
a_segundo_palo = np.arctan(m_segundo_palo) * 180 / np.pi

desvio_para_gol = a_primer_palo - A_PELOTA_CORNER * 180 / np.pi , (a_segundo_palo - A_PELOTA_CORNER * 180 / np.pi)

print(f"El angulo entre el jugador y el primer palo es de {a_primer_palo:.2f} grados")
print(f"El angulo entre el jugador y el segundo palo es de {a_segundo_palo:.2f} grados")
print(f"Condicion de gol: cualquier desvio aplicado del jugador entre {desvio_para_gol[0]:.2f} y {desvio_para_gol[1]:.2f} grados (intervalo abierto)")

```
