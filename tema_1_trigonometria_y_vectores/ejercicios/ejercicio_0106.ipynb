# Ejercicio 6, Desplazamiento en 3D, Trabajo con vectores.

Un habitacion de 3m x 4m x 5m. Un mosca que sale de una esquina termina su vuelo en la esquina 'diametralmente' opuesta. 

- (a) Halla el vector desplazamiento del desplazamiento de la mosca con i, j, k.
- (b) Cual es el desplazamiento total de la mosca?
- (c) si la mosca camina en vez de volar, cual es el desplazamiento total minimo que puede hacer?

![](ejercicio_0106.png)

```python
import numpy as np

# parte a)
m_3d = 3, 4, 5

print(f"vector m: {m_3d}")

# parte b)
print(f"Si la mosca vuela en linea recta, la distancia recorrida es:")

d_m_3d  = np.sqrt(np.sum(np.array(m_3d)**2))

print(f"{d_m_3d:.2f}m")

# parte c) la mosca camina diagonalmente y despues hacia arriba.
print(f"Si la mosca camina diagonalmente y despues hacia arriba, la distancia recorrida es:")

m_2d = 4, 5
d_m_2d  = np.sqrt(np.sum(np.array(m_2d)**2))

print(f"caminata diagonal: {d_m_2d:.2f}m")

m1 = 0, 4, 5
m2 = 3, 4, 5
m1_m2 = np.array(m2) - np.array(m1)
d_m1_m2 = np.sqrt(np.sum(m1_m2**2))

print(f"caminata hacia arriba: {d_m1_m2:.2f}m")
print(f"distancia caminada total: {(d_m_2d + d_m1_m2):.2f}m")
```
