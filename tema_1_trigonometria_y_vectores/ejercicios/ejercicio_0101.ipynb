# Ejercicio 1: Una idea notable!

Erastostenes midio por primera vez la circunferencia de la Tierra, en el solsticio de verano. Los rayos del sol caen verticalmente sobre Siena, pero en Alejandría, Egipto, forman un ángulo de 7.2 grados con la vertical. 

Erastostenes sabía que la distancia entre Siena y Alejandría era de aproximadamente 800 km. Utilizando la diferencia de ángulos, Erastostenes calculó la circunferencia de la Tierra.

![](ejercicio_0101.png)


```python
import numpy as np

PI = np.pi
ALPHA = 7.2
X = 800

sin = np.sin
alpha_rad = ALPHA * PI / 180

# SOH
h = X / sin(alpha_rad)

# Circunferencia de la Tierra
cfa_tierra = (2 * PI * h).round(2)
# Dato: La circunferencia de la Tierra es de 40075 km
err_rel = ((np.abs(cfa_tierra - 40075) / 40075) * 100).round(2)

print(f"La estimacion de la cfa de la Tierra es de {cfa_tierra} km")
print(f"Error relativo: {err_rel}%")
```
