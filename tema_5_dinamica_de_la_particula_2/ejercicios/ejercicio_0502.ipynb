# Ejercicio 2, Desigualdad de Newton $|F_s| \leq \mu N$

Una caja tiene 2500 Kg de masa y descansa sobre una superficie horizontal. El coeficiente de fricción estática entre la caja y la superficie es de 0.65.

La caja comprime un resorte de constante $ k = 64300 N/m $, ubicado horizontalmente entra la esta y la pared.

El resorte es comprimido 0.02 m y la caja se suelta. Que fuerza de rozamiento actúa sobre la caja, verificar si se cumple la desigualdad de Newton $|F_s| \leq \mu N$.

```python
import sympy as sp

# Datos
## Masa caja
m = 2500 # kg
## Coeficiente de fricción
u = 0.65
## Aceleración de la gravedad
g = 9.81 # m/s^2
## Constante k, elastica
k_0 = 64300 # N/m
## Compresión del resorte
x_0 = -0.02 # m

# Variables
F_r, F_e, k, x = sp.symbols('F_r F_e k x_0')

# Analisis de fuerzas eje x
print('Analisis de fuerzas eje x')
nw_x = sp.Eq(F_e - F_r,  0)
display(nw_x)
nw_x = nw_x.subs(F_e, - k*x)
display(nw_x)
nw_x = sp.Eq(F_r, sp.solve(nw_x, F_r)[0])
display(nw_x)

# Solución
print('Solución')
nw_x = nw_x.subs({k: k_0, x: x_0})
display(nw_x)
```
