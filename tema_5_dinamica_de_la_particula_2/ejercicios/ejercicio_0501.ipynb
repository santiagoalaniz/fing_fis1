# Ejercicio 1, Equilibrio de Fuerzas.

El bloque B de la figura pesa 712 N, el coef. de rozamiento entre el bloque B y la mesa es de 0.25. Halle el peso maximo del bloque A para que el sistema este en equilibrio.

```python
import sympy as sp
import numpy as np

# Datos
## Fuerza peso en el bloque B
mod_P = 712 # N
## Coeficiente de friccion estatica de la mesa con el bloque B
mu_s = 0.25 # adimensional
## Angulo de la Tension respect a la horizontal del bloque B y vertical del bloque A
theta_T = 41 * np.pi / 180 # radianes
## Fr = mu_s * N, N = -P
mod_Fr = mod_P * mu_s

# Variables
T, P_A, P_B, Fr, theta = sp.symbols('T P_A P_B Fr theta')

# Analisis de fuerzas del bloque B en el eje x
print('Bloque B')
## Sumatoria de fuerzas en x
eq_x = sp.Eq(T * sp.cos(theta) - Fr, 0)
display(eq_x)
# Despeje T, usando eq_x
T_sol = sp.solve(eq_x, T)[0]
eq_x = sp.Eq(T, T_sol)
display(eq_x)

# Analisis de fuerzas del bloque A en el eje y
print('Bloque A')
## Sumatoria de fuerzas en x
eq_y = sp.Eq(T * sp.sin(theta) - P_A, 0)
display(eq_y)
# Despeje P_A, usando eq_y, sustituyendo T por su valor
P_A_sol = sp.solve(eq_y.subs(T, T_sol), P_A)[0]
eq_y = sp.Eq(P_A, P_A_sol)
display(eq_y)

print('Solucion')
# valores utiles
display(sp.Eq(Fr, mod_Fr))
display(sp.Eq(theta, 41))
## Sustitucion de valores
T_sol = T_sol.subs([(theta, theta_T), (Fr, mod_Fr)])
P_A_sol = P_A_sol.subs([(theta, theta_T), (Fr, mod_Fr)])
eq_y = eq_y.subs([(theta, theta_T), (Fr, mod_Fr)]).evalf(4)
display(eq_y)
```
