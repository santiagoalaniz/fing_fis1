# Ejercicio 4, Efecto al disminuir la Fuerza Normal.

En la figura, A es un bloque de masa $M_A$ y B es un bloque de masa $M_B$. 

Los coeficientes de rozamiento $\mu_{s}, \mu_{k}$, es $0.18$ y $0.15$, respectavimente.

- (a) Determine la masa mínima del bloque C que debe colocarse sobre A para evitar que éste deslice. 

- (b) El bloque C es levantado súbitamente de A. ¿Cuál es la aceleración del bloque A?


```python
import sympy as sp

# Datos
## Masas de los bloqueas A y B.
m_A, m_B = sp.symbols('M_A M_B')
## Aceleración de la gravedad.
g = sp.symbols('g')
## Coeficientes estaticos y cinéticos de fricción.
mu_s, mu_k = sp.symbols('mu_s mu_k')

# Variables
## Masas del bloque C.
m_C = sp.symbols('M_C')
## Fuerza Tension
T = sp.symbols('T')
## Fuerza de fricción estatica y cinética.
F_s, F_k = sp.symbols('F_s F_k')
## Fuerza normales de los bloques A y B.
N_A, N_B, N_AC = sp.symbols('N_A N_B N_AC')
## Aceleracion
a = sp.symbols('a')

# Analisis de fuerzas del bloque A eje x.
print('Analisis eje x de fuerzas del bloque A estatico')
nw_eq_x = sp.Eq(T - F_s, 0)
display(nw_eq_x)
## T como funcion de caja B
t_eq = sp.Eq(T, m_B*g)
display(t_eq)
## F_s como funcion de N_AC
f_s_eq = sp.Eq(F_s, mu_s*N_AC)
display(f_s_eq)
## N_AC como funcion de m_C Y m_A
n_ac_eq = sp.Eq(N_AC, m_C*g + m_A*g)
display(n_ac_eq)

# Sustitución de ecuaciones
print('Sustitución de ecuaciones')
## Substitucion de T
nw_eq_x = nw_eq_x.subs(T, t_eq.rhs)
display(nw_eq_x)
## Substitucion de F_s y N_AC
nw_eq_x = nw_eq_x.subs(F_s, f_s_eq.rhs.subs(N_AC, n_ac_eq.rhs))
display(nw_eq_x)

# Solución A
print('Solución A')
## Despeje de M_C
m_c_eq = sp.Eq(m_C, sp.solve(nw_eq_x, m_C)[0])
display(m_c_eq)

# Analisis de fuerzas del bloque A eje x, cuando se levanta el bloque C y este se desliza.
print('Analisis eje x de fuerzas del bloque A acelerado cuando se levanta el bloque C')
nw_eq_x = sp.Eq(T - F_k, m_A*a)
display(nw_eq_x)
## T como funcion de caja B
t_eq = sp.Eq(T, m_B*g - m_B*a)
display(t_eq)
## F_k como funcion de N_A
f_k_eq = sp.Eq(F_k, mu_k*N_A)
display(f_k_eq)
## N_A como funcion de m_A
n_a_eq = sp.Eq(N_A, m_A*g)
display(n_a_eq)

# Sustitución de ecuaciones
print('Sustitución de ecuaciones')
## Substitucion de T como funcion de m_B y a
nw_eq_x = nw_eq_x.subs(T, t_eq.rhs)
display(nw_eq_x)
## Substitucion de F_k y N_A
nw_eq_x = nw_eq_x.subs(F_k, f_k_eq.rhs.subs(N_A, n_a_eq.rhs))
display(nw_eq_x)

# Solución B
print('Solución B')
## Despeje de a
a_eq = sp.Eq(a, sp.solve(nw_eq_x, a)[0])
display(a_eq)
```
