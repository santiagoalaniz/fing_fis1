# Ejercicio 3, Trabajo con $m_1$ como unidad de masa y $\vec{T_3}$ como parámetro conocido.

Tres bloques estan unidos como muestra la figura, sobre una mesa horizontal carente de friccion.

![Ejercicio 3](ejercicio_0403.png)

Son tirados hacia la derecha con una fuerza externa $\vec{T_3}$. Si $m_2 = \alpha m_1$ y $m_3 = \beta m_1$, siendo $\alpha$ y $\beta$ constantes adimensionales.

Se pide:

- (a) La aceleración del sistema
- (b) Las tensiones $\vec{T_1}$ y $\vec{T_2}$

# Solucion

## (a) La aceleración del sistema

Solo hay una fuerza externa actuando sobre el sistema, la fuerza $\vec{T_3}$, por lo que la aceleración del sistema es:

$$a = \frac{F_{ext}}{M} = \frac{T_3}{m_1 + m_2 + m_3} = \frac{T_3}{m_1 + \alpha m_1 + \beta m_1} = \frac{T_3}{(1 + \alpha + \beta)m_1}$$

## (b) Las tensiones $\vec{T_1}$ y $\vec{T_2}$

**importante**: notar que la acelaración del sistema es la misma para todos los bloques, podemos usar esta informacion mas el analisis de cuerpo libre para definir las tensiones.

$$
\begin{align*}
    \text{ (bloque 3) }  m_3 a &= T_3 - T_2  \\ 
    \text{ (bloque 2) }  m_2 a &= T_2 - T_1  \\
    \text{ (bloque 1) }  m_1 a &= T_1            
\end{align*}
$$

Notar como sumar las tres ecuaciones anteriores se obtiene la aceleración del sistema.

El valor de $T_1$ es una substitución directa de la aceleración.

$$T_1 = m_1 a = \frac{m_1 T_3}{(1 + \alpha + \beta)m_1} = \frac{T_3}{1 + \alpha + \beta}$$

Para encontrar $T_2$, se puede substituir $T_1$ en la ecuación del bloque 2.

$$T_2 = m_2 a + \frac{T_3}{1 + \alpha + \beta} = \frac{\alpha T_3}{1 + \alpha + \beta} + \frac{T_3}{1 + \alpha + \beta} = \frac{\alpha + 1}{1 + \alpha + \beta} T_3$$
