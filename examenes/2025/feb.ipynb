## Ejercicio 1

Una pelota se deja caer desde una altura de 19.6 m. Al rebotar contra el suelo, la pelota sale con 
una velocidad cuyo módulo es la mitad de la velocidad que tenía inmediatamente antes de llegar al 
piso. ¿Cuánto tiempo transcurre entre el primer y el segundo rebote en el piso?

```python
import sympy as sp
from IPython.display import Markdown

# Definir símbolos
x0, v0, a0, t = sp.symbols('x0 v0 a0 t')

# Constantes
X0, V0, A0 = 19.6, 0, -9.81  # Posición inicial, velocidad inicial y aceleración debida a la gravedad

# Definir la ecuación de movimiento
x_t = x0 + v0*t + sp.Rational(1, 2) * a0 * t**2
x_t = x_t.subs(a0, A0)  # Sustituir a0 por g
x_t = x_t.subs(v0, V0)  # Sustituir v0 por V0
x_t = x_t.subs(x0, X0)  # Sustituir x0 por X0
display(Markdown(f"**Ecuación de movimiento:** $x(t) = {sp.latex(x_t)}$"))

# t_imp, i.e tiempo de impacto en el suelo
t_imp = sp.solve(x_t, t)[1]
display(Markdown(f"**Tiempo de impacto:** $t_{{imp}} = {sp.latex(t_imp)}$"))

# velocidad en el impacto
v_t = sp.diff(x_t, t)
v_t = v_t.subs(t, t_imp)  # Sustituir t por t_imp
display(Markdown(f"**Velocidad después del rebote:** $v(t_{{imp}} - \\Delta) = {sp.latex(v_t)}$"))

# rebote, con la mitad de la velocidad pero, en sentido contrario
v_t = -v_t / 2
display(Markdown(f"**Velocidad después del rebote:** $v(t_{{imp}} + \\Delta) = {sp.latex(v_t)}$"))

Ximp, Vimp, Aimp = 0, v_t, A0  # Posición inicial, velocidad inicial y aceleración debida a la gravedad
# Definir la ecuación de movimiento después del rebote
x_t = x0 + v0*t + sp.Rational(1, 2) * a0 * t**2
x_t = x_t.subs(a0, Aimp)  # Sustituir a0 por g
x_t = x_t.subs(v0, Vimp)  # Sustituir v0 por Vimp
x_t = x_t.subs(x0, Ximp)  # Sustituir x0 por Ximp
display(Markdown(f"**Ecuación de movimiento después del rebote:** $x(t) = {sp.latex(x_t)}$"))
# t_imp, i.e tiempo de impacto en el suelo
t_imp_2 = sp.solve(x_t, t)[1]
display(Markdown(f"**Tiempo de impacto después del rebote:** $t_{{imp}} = {sp.latex(t_imp_2)}$"))
```

## Ejercicio 2

Un disparador de pelotitas de golf, ubicado a una altura $h = 10\, \text{m}$, forma un ángulo de $30^\circ$ con la horizontal y lanza las pelotitas a una velocidad inicial de módulo $v_0 = 2.0\, \text{m/s}$.
Desde el piso, directamente debajo del disparador, una camioneta de juguete que parte desde el reposo se mueve con aceleración constante.
¿Cuál debe ser el módulo de dicha aceleración para que la camioneta atrape la pelotita, asumiendo que ambos comienzan a moverse simultáneamente?



```python
import sympy as sp
from IPython.display import Markdown

# Datos
h = 10 # Altura inicial
v0_mod = 2 # modulo de la velocidad inicial 
alpha = 30 # Ángulo de lanzamiento en grados
g = -9.81  # Aceleración debida a la gravedad

# Paso 1 separar la velocidad inicial en sus componentes
v0_x = v0_mod * sp.cos(sp.rad(alpha))  # Componente horizontal
v0_y = v0_mod * sp.sin(sp.rad(alpha))  # Componente vertical

display(Markdown(
  f"**Componentes de la velocidad inicial:** $v_{{0x}} = \
  {sp.latex(v0_x)},\\quad v_{{0y}} = {sp.latex(v0_y)}$"
))

# Paso 2 calcular cuando la pelota alcanza el suelo Xy(t) = 0
t = sp.symbols('t')
Xy_0 = h # Posición inicial en y
Vy_0 = v0_y # Velocidad inicial en y
Ay_0 = g
# Ecuación de movimiento en y
Xy_t = Xy_0 + Vy_0 * t + sp.Rational(1, 2) * Ay_0 * t**2
display(Markdown(f"**Ecuación de movimiento en y:** $X_y(t) = {sp.latex(Xy_t)}$"))
# Resolver para t cuando Xy(t) = 0
t_imp = sp.solve(Xy_t, t)[1]  # Tomamos la solución positiva
display(Markdown(f"**Tiempo de impacto:** $t_{{imp}} = {sp.latex(t_imp)}$"))

# Paso 3 calcular la distancia horizontal recorrida hasta el impacto
# Funcion de la distancia horizontal recorrida
Xx_t = v0_x * t  # Movimiento horizontal con velocidad constante
display(Markdown(f"**Ecuación de movimiento horizontal:** $X_x(t) = {sp.latex(Xx_t)}$"))
Xx_t_pelota = float(v0_x * t_imp) # Distancia horizontal recorrida
display(Markdown(f"**Distancia horizontal recorrida hasta el impacto:** $X_x(t_{{imp}}) = {sp.latex(Xx_t_pelota)}$ m"))

# Paso 4, movimiento de la camioneta, se mueve con aceleración constante
# Ecuación de movimiento de la camioneta
Xx_0 = 0  # Posición inicial de la camioneta
Vx_0 = 0  # Velocidad inicial de la camioneta
Ax_0 = sp.symbols('a')  # Aceleración de la camioneta (desconocida)
Xx_t = Xx_0 + Vx_0 * t + sp.Rational(1, 2) * Ax_0 * t**2
display(Markdown(f"**Ecuación de movimiento de la camioneta:** $X_x(t) = {sp.latex(Xx_t)}$"))

# Paso 5, igualar las posiciones de la pelota y la camioneta en el tiempo de impacto
eq = sp.Eq(Xx_t.subs(t, t_imp), Xx_t_pelota)
# Resolver para la aceleración de la camioneta
a_camioneta = sp.solve(eq, Ax_0)[0]
display(Markdown(f"**Aceleración de la camioneta:** $a = {a_camioneta:.1f}\\ \\mathrm{{m/s^2}}$"))
```

## Ejercicio 3

Considera el sistema de masas y poleas indicado en la figura. El bloque $B$, de masa $m$, desciende verticalmente moviendo el sistema. 

El contacto entre el bloque $A$, de masa $2m$ y la superficie horizontal es liso. Las poleas no tienen masa ni fricción en el eje y el hilo es ideal.

Calcula el módulo de la aceleración del bloque $A$.

```python
import sympy as sp
from IPython.display import Markdown

# ✏️ Definir símbolos
a_A, a_B, T, g, m = sp.symbols('a_A a_B T g m')

# Paso 1: Relación entre aceleraciones (por conservación de cuerda)
eq_acc = sp.Eq(a_B, sp.Rational(2, 3) * a_A)
display(
  Markdown("**Paso 1:** Relación entre aceleraciones por conservación de cuerda"),
  sp.Eq(sp.Eq(3 * a_B - 2 * a_A, 0), eq_acc)
)

# Paso 2: Segunda ley de Newton para el bloque A
eq_A = sp.Eq(T, m * a_A)
display(
  Markdown("**Paso 2:** Segunda ley de Newton para el bloque A"),
  eq_A
)

# Paso 3: Segunda ley de Newton para el bloque B
eq_B = sp.Eq(m * g - 3 * T, m * a_B)
display(
  Markdown("**Paso 3:** Segunda ley de Newton para el bloque B"),
  eq_B
)

# Paso 4: Sustituir (1) y (2) en la ecuación del bloque B
eq_B_sub = eq_B.subs({a_B: eq_acc.rhs, T: eq_A.rhs})
display(
  Markdown("**Paso 4:** Sustitución de (1) y (2) en la ecuación del bloque B"),
  eq_B_sub
)

# Paso 5: Resolver para a_A
sol_aA = sp.solve(eq_B_sub, a_A)[0]
display(
  Markdown("**Paso 5:** Resolución de la ecuación para $a_A$"),
  sp.Eq(a_A, sol_aA)
)

# Paso 6: Calcular a_B y T
sol_aB = eq_acc.subs(a_A, sol_aA).rhs
sol_T = eq_A.subs(a_A, sol_aA).rhs
display(
  Markdown("**Paso 6:** Cálculo de $a_B$ y $T$"),
  sp.Eq(a_B, sol_aB),
  sp.Eq(T, sol_T)
)
```

## Ejercicio 4

El sistema de la figura consta de tres cuerdas ideales unidas por un nudo sin masa: una cuerda está unida a la pared y se mantiene horizontal, otra cuerda sostiene un bloque de masa $m_c$ y la tercera pasa por una polea ideal y sin masa y está unida en su otro extremo a un bloque de masa $m_a$. Dicho bloque está apoyado en una superficie inclinada un ángulo $\theta = 45^\circ$ respecto de la horizontal. Existe fricción entre el bloque y dicha superficie, con coeficiente de fricción estática $\mu_s$.
¿Cuál es la masa $m_c$ máxima que mantiene el sistema en equilibrio?


```python
import sympy as sp
from IPython.display import Markdown, display

# Parámetros físicos
theta_deg = 45
theta = sp.rad(theta_deg)
g = 9.81

# Variables simbólicas
mc, ma, mu_s = sp.symbols('mc ma mu_s')
T1, T2, T3 = sp.symbols('T1 T2 T3')

# Fuerzas en el bloque colgante
F_peso = mc * g                  # Peso
eq_colgante = sp.Eq(T1 - F_peso, 0)  # Tensión equilibra el peso
display(Markdown("### Ecuación bloque colgante:"),
        Markdown("**Eje y vertical**"),
        eq_colgante)

# Fuerzas en el bloque sobre el plano inclinado
F_peso = ma * g * sp.sin(theta)
F_normal = ma * g * sp.cos(theta)
F_friccion = mu_s * F_normal
eq_plano = sp.Eq(T2 - F_peso - F_friccion, 0)  # Tensión sube por el plano
display(Markdown("### Ecuacion bloque sobre plano inclinado:"),
        Markdown("**Eje x rotado a 45°**"),
        eq_plano)

# Equilibrio en el nudo sin masa
# Descomposición de T2 a 45° (es la que va hacia la polea)
T2x = T2 * sp.cos(theta)
T2y = T2 * sp.sin(theta)
eq_nudo_x = sp.Eq(T3, T2x)
eq_nudo_y = sp.Eq(T1, T2y)
display(Markdown("### Ecuaciones de equilibrio en el nudo:"),
        Markdown("**Son dos porque el nudo pued acelerar en ambas direcciones**"),
        eq_nudo_x, eq_nudo_y)

# Mostrar ecuaciones
display(Markdown("### Ecuaciones de equilibrio:"),
        eq_colgante, eq_plano, eq_nudo_x, eq_nudo_y)

# Sustituciones para eliminar tensiones
# Paso 1: usar eq_nudo_y -> T1 = T2 * sin(theta)
# Paso 2: usar eq_colgante -> T1 = mc * g
# Paso 3: igualar y despejar mc

expr_mc = sp.solve(eq_nudo_y.subs(T1, mc * g), T2)[0]  # T2 en función de mc
expr_T2 = sp.solve(eq_plano.subs(T2, expr_mc), mc)[0]  # mc en función de ma, mu_s

# Resultado final
display(Markdown(f"### Resultado:"))
display(sp.simplify(expr_T2))
```

## Ejercicio 5

Un futbolista tiene una velocidad horizontal de 1 m/s con respecto al suelo en el momento en que recibe con el pecho una pelota que viaja en su misma dirección y sentido. Inmediatamente después de rebotar en el pecho del futbolista, la pelota tiene una velocidad horizontal de sentido contrario al inicial y módulo 4 m/s con respecto al suelo. Si suponemos que la masa de la pelota es mucho, mucho menor que la del futbolista y que el choque es completamente elástico, ¿cuál era el módulo de la velocidad de la pelota con respecto al suelo inmediatamente antes del impacto?


```python
import sympy as sp
from IPython.display import Markdown, display

# Definir símbolos
V_f = sp.symbols('v_f')  # Velocidad del futbolista respecto al suelo
V_p2 = sp.symbols('v_p2')  # Velocidad de la pelota después del choque respecto al suelo
V_p1 = sp.symbols('v_p1')  # Velocidad de la pelota antes del choque respecto al suelo
V_p2_f = sp.symbols('v_{p2/f}')  # Velocidad de la pelota respecto al futbolista después del choque
V_p1_f = sp.symbols('v_{p1/f}')  # Velocidad de la pelota antes del choque respecto al futbolista

# Ecuacion de relacion entre velocidades
## V_x_y = V_x - V_y, donde:
## V_x es la velocidad del objeto X respecto a un sistema de referencia
## V_y es la velocidad del objeto Y respecto al mismo sistema de referencia
## V_x_y es la velocidad del objeto X respecto al objeto Y
eq_V_p2_f = sp.Eq(V_p2_f, V_p2 - V_f)
display(Markdown("### Ecuación de relación entre velocidades despues del choque:"),
        eq_V_p2_f)

# Aplicacion de datos de la realidad a la ecuación
## Velocidad del futbolista respecto al sistema de referencia (suelo)
eq_V_p2_f = eq_V_p2_f.subs(V_f, 1)
display(
  Markdown
    ("#### *...Un futbolista tiene una velocidad horizontal de 1 m/s con respecto al suelo...*"),
  sp.Eq(V_f, 1),
  eq_V_p2_f)

## Velocidad de la pelota después del choque respecto al sistema de referencia (suelo)
eq_V_p2_f = eq_V_p2_f.subs(V_p2, - 4) ## direccion opuesta al futbolista
display(
  Markdown(
    "#### *Inmediatamente después de rebotar en el pecho del futbolista, la pelota tiene una velocidad horizontal de sentido contrario al inicial y módulo 4 m/s con respecto al suelo*"
  ),
  sp.Eq(V_p2, -4),
  eq_V_p2_f
)
## Aplicamos condicion de igualdad por choque elastico
eq_V_p1_f = sp.Eq(
  sp.Eq(V_p1_f, V_p1 - V_f),  # Velocidad de la pelota antes del choque respecto al futbolista
  sp.Eq(- V_p2_f, - eq_V_p2_f.rhs), # Choque elastico, conservacion del modulo de la velocidad con sentido opuesto
)
display(Markdown("### Choque elastico, conservacion del modulo de la velocidad con sentido opuesto"),
        eq_V_p1_f)
# Incorporamos el valor de V_f otra vez
eq_V_p1_f = eq_V_p1_f.subs(V_f, 1)
display(
  Markdown("#### *...El futbolista se mueve con una velocidad horizontal de 1 m/s con respecto al suelo...*"),
  sp.Eq(V_f, 1),
  eq_V_p1_f
)
## Finalmente, despejamos la velocidad de la pelota antes del choque
V_p1_f_val = -eq_V_p2_f.rhs  # Esto es -(-4 - 1) = 5
eq_p1 = sp.Eq(V_p1 - 1, V_p1_f_val)  # Sustituimos V_f = 1

# Mostrar resultado
display(
    Markdown("#### Despeje de la velocidad de la pelota antes del impacto:"),
    eq_p1,
    sp.Eq(V_p1, sp.solve(eq_p1, V_p1)[0])
)
```

## Ejercicio 6

Dos niños juegan en un sube y baja, que consiste en una tabla de largo $L = 3{,}0\,m$ que puede rotar libremente alrededor de un eje que pasa por su punto medio $O$. El niño más pequeño, de masa $m = 20\,kg$ se queda sentado en el extremo de la tabla que está apoyado contra el piso. El otro niño, de masa $M = 30\,kg$, comienza a caminar por la tabla, desde la posición del primer niño, buscando llegar al otro extremo. ¿Qué distancia $x$ habrá recorrido antes de que la tabla comience a moverse?


```python
import sympy as sp
from IPython.display import Markdown, display

# Definir símbolos
m = sp.symbols('m')   # Masa del ninio pequeño
M = sp.symbols('M')   # Masa del ninio grande
g = sp.symbols('g')   # Aceleración debida a la gravedad

# Datos del problema
dato_m = 20           # Masa del ninio pequeño en kg
dato_M = 30           # Masa del ninio grande en kg
dato_g = 9.81         # Aceleración debida a la gravedad en m/s²
dato_xm = 1.5         # Distancia del ninio pequeño al eje de rotación en m

# Concepto relacionado con el ejercicio, torque
display(Markdown("### Concepto relacionado: Torque"),
       Markdown("**Torque:** $\\tau = r \\cdot F$ donde $r$ es la distancia al eje de rotación y " \
       "$F$ es la fuerza aplicada."))

# Torque del ninio pequeño, sentado en el extremo de la tabla
tau_m = sp.symbols('tau_m')  # Torque del ninio pequeño
eq_tau_m = sp.Eq(tau_m, m * g * dato_xm)
display(Markdown("### Torque del ninio pequeño:"),
       eq_tau_m)

# Torque del ninio grande, que camina por la tabla
tau_M = sp.symbols('tau_M')  # Torque del ninio grande
var_xM = sp.symbols('x_M')  # Distancia del ninio grande al eje de rotación
eq_tau_M = sp.Eq(tau_M, M * g * (var_xM - dato_xm))
display(Markdown("### Torque del ninio grande, que camina por la tabla:"),
       eq_tau_M)

# Igualar los torques
eq_tau = sp.Eq(eq_tau_m.rhs, eq_tau_M.rhs)
display(Markdown("### Igualar los torques para encontrar la distancia recorrida por el ninio grande" \
       "hasta que se mueva:"),
       eq_tau)

# Despejar la distancia recorrida por el ninio grande
x_M = sp.solve(eq_tau, var_xM)[0]
var_xM = sp.Eq(var_xM, x_M)
display(Markdown("### Despeje de la distancia recorrida por el ninio grande hasta que se mueva:"),
       var_xM,
       var_xM.subs({m: dato_m, M: dato_M, g: dato_g}))
```
