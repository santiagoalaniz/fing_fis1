## Ejercicio 1

Una pelota se deja caer desde una altura de 19.6 m. Al rebotar contra el suelo, la pelota sale con 
una velocidad cuyo módulo es la mitad de la velocidad que tenía inmediatamente antes de llegar al 
piso. ¿Cuánto tiempo transcurre entre el primer y el segundo rebote en el piso?

```python
import sympy as sp
from IPython.display import Markdown

# Definir símbolos
x0, v0, a0, t = sp.symbols('x0 v0 a0 t')

# Constantes
X0, V0, A0 = 19.6, 0, -9.81  # Posición inicial, velocidad inicial y aceleración debida a la gravedad

# Definir la ecuación de movimiento
x_t = x0 + v0*t + sp.Rational(1, 2) * a0 * t**2
x_t = x_t.subs(a0, A0)  # Sustituir a0 por g
x_t = x_t.subs(v0, V0)  # Sustituir v0 por V0
x_t = x_t.subs(x0, X0)  # Sustituir x0 por X0
display(Markdown(f"**Ecuación de movimiento:** $x(t) = {sp.latex(x_t)}$"))

# t_imp, i.e tiempo de impacto en el suelo
t_imp = sp.solve(x_t, t)[1]
display(Markdown(f"**Tiempo de impacto:** $t_{{imp}} = {sp.latex(t_imp)}$"))

# velocidad en el impacto
v_t = sp.diff(x_t, t)
v_t = v_t.subs(t, t_imp)  # Sustituir t por t_imp
display(Markdown(f"**Velocidad después del rebote:** $v(t_{{imp}} - \\Delta) = {sp.latex(v_t)}$"))

# rebote, con la mitad de la velocidad pero, en sentido contrario
v_t = -v_t / 2
display(Markdown(f"**Velocidad después del rebote:** $v(t_{{imp}} + \\Delta) = {sp.latex(v_t)}$"))

Ximp, Vimp, Aimp = 0, v_t, A0  # Posición inicial, velocidad inicial y aceleración debida a la gravedad
# Definir la ecuación de movimiento después del rebote
x_t = x0 + v0*t + sp.Rational(1, 2) * a0 * t**2
x_t = x_t.subs(a0, Aimp)  # Sustituir a0 por g
x_t = x_t.subs(v0, Vimp)  # Sustituir v0 por Vimp
x_t = x_t.subs(x0, Ximp)  # Sustituir x0 por Ximp
display(Markdown(f"**Ecuación de movimiento después del rebote:** $x(t) = {sp.latex(x_t)}$"))
# t_imp, i.e tiempo de impacto en el suelo
t_imp_2 = sp.solve(x_t, t)[1]
display(Markdown(f"**Tiempo de impacto después del rebote:** $t_{{imp}} = {sp.latex(t_imp_2)}$"))
```

# Ejercicio 2

Un disparador de pelotitas de golf, ubicado a una altura $h = 10\, \text{m}$, forma un ángulo de $30^\circ$ con la horizontal y lanza las pelotitas a una velocidad inicial de módulo $v_0 = 2.0\, \text{m/s}$.
Desde el piso, directamente debajo del disparador, una camioneta de juguete que parte desde el reposo se mueve con aceleración constante.
¿Cuál debe ser el módulo de dicha aceleración para que la camioneta atrape la pelotita, asumiendo que ambos comienzan a moverse simultáneamente?



```python
import sympy as sp
from IPython.display import Markdown

# Datos
h = 10 # Altura inicial
v0_mod = 2 # modulo de la velocidad inicial 
alpha = 30 # Ángulo de lanzamiento en grados
g = -9.81  # Aceleración debida a la gravedad

# Paso 1 separar la velocidad inicial en sus componentes
v0_x = v0_mod * sp.cos(sp.rad(alpha))  # Componente horizontal
v0_y = v0_mod * sp.sin(sp.rad(alpha))  # Componente vertical

display(Markdown(
  f"**Componentes de la velocidad inicial:** $v_{{0x}} = \
  {sp.latex(v0_x)},\\quad v_{{0y}} = {sp.latex(v0_y)}$"
))

# Paso 2 calcular cuando la pelota alcanza el suelo Xy(t) = 0
t = sp.symbols('t')
Xy_0 = h # Posición inicial en y
Vy_0 = v0_y # Velocidad inicial en y
Ay_0 = g
# Ecuación de movimiento en y
Xy_t = Xy_0 + Vy_0 * t + sp.Rational(1, 2) * Ay_0 * t**2
display(Markdown(f"**Ecuación de movimiento en y:** $X_y(t) = {sp.latex(Xy_t)}$"))
# Resolver para t cuando Xy(t) = 0
t_imp = sp.solve(Xy_t, t)[1]  # Tomamos la solución positiva
display(Markdown(f"**Tiempo de impacto:** $t_{{imp}} = {sp.latex(t_imp)}$"))

# Paso 3 calcular la distancia horizontal recorrida hasta el impacto
# Funcion de la distancia horizontal recorrida
Xx_t = v0_x * t  # Movimiento horizontal con velocidad constante
display(Markdown(f"**Ecuación de movimiento horizontal:** $X_x(t) = {sp.latex(Xx_t)}$"))
Xx_t_pelota = float(v0_x * t_imp) # Distancia horizontal recorrida
display(Markdown(f"**Distancia horizontal recorrida hasta el impacto:** $X_x(t_{{imp}}) = {sp.latex(Xx_t_pelota)}$ m"))

# Paso 4, movimiento de la camioneta, se mueve con aceleración constante
# Ecuación de movimiento de la camioneta
Xx_0 = 0  # Posición inicial de la camioneta
Vx_0 = 0  # Velocidad inicial de la camioneta
Ax_0 = sp.symbols('a')  # Aceleración de la camioneta (desconocida)
Xx_t = Xx_0 + Vx_0 * t + sp.Rational(1, 2) * Ax_0 * t**2
display(Markdown(f"**Ecuación de movimiento de la camioneta:** $X_x(t) = {sp.latex(Xx_t)}$"))

# Paso 5, igualar las posiciones de la pelota y la camioneta en el tiempo de impacto
eq = sp.Eq(Xx_t.subs(t, t_imp), Xx_t_pelota)
# Resolver para la aceleración de la camioneta
a_camioneta = sp.solve(eq, Ax_0)[0]
display(Markdown(f"**Aceleración de la camioneta:** $a = {a_camioneta:.1f}\\ \\mathrm{{m/s^2}}$"))
```

## Ejercicio 3

Considera el sistema de masas y poleas indicado en la figura. El bloque $B$, de masa $m$, desciende verticalmente moviendo el sistema. 

El contacto entre el bloque $A$, de masa $2m$ y la superficie horizontal es liso. Las poleas no tienen masa ni fricción en el eje y el hilo es ideal.

Calcula el módulo de la aceleración del bloque $A$.

```python
import sympy as sp
from IPython.display import Markdown

# ✏️ Definir símbolos
a_A, a_B, T, g, m = sp.symbols('a_A a_B T g m')

# Paso 1: Relación entre aceleraciones (por conservación de cuerda)
eq_acc = sp.Eq(a_B, sp.Rational(2, 3) * a_A)
display(
  Markdown("**Paso 1:** Relación entre aceleraciones por conservación de cuerda"),
  sp.Eq(sp.Eq(3 * a_B - 2 * a_A, 0), eq_acc)
)

# Paso 2: Segunda ley de Newton para el bloque A
eq_A = sp.Eq(T, m * a_A)
display(
  Markdown("**Paso 2:** Segunda ley de Newton para el bloque A"),
  eq_A
)

# Paso 3: Segunda ley de Newton para el bloque B
eq_B = sp.Eq(m * g - 3 * T, m * a_B)
display(
  Markdown("**Paso 3:** Segunda ley de Newton para el bloque B"),
  eq_B
)

# Paso 4: Sustituir (1) y (2) en la ecuación del bloque B
eq_B_sub = eq_B.subs({a_B: eq_acc.rhs, T: eq_A.rhs})
display(
  Markdown("**Paso 4:** Sustitución de (1) y (2) en la ecuación del bloque B"),
  eq_B_sub
)

# Paso 5: Resolver para a_A
sol_aA = sp.solve(eq_B_sub, a_A)[0]
display(
  Markdown("**Paso 5:** Resolución de la ecuación para $a_A$"),
  sp.Eq(a_A, sol_aA)
)

# Paso 6: Calcular a_B y T
sol_aB = eq_acc.subs(a_A, sol_aA).rhs
sol_T = eq_A.subs(a_A, sol_aA).rhs
display(
  Markdown("**Paso 6:** Cálculo de $a_B$ y $T$"),
  sp.Eq(a_B, sol_aB),
  sp.Eq(T, sol_T)
)
```
