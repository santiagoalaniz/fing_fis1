## Ejercicio 1

Un proyectil parte desde el suelo con velocidad inicial $\vec{v}_0$. Calcule el ángulo $\theta$ que debe tener la velocidad inicial, para que luego de superar el obstáculo que se encuentra a una distancia $D/2$ el proyectil impacte en el punto P mostrado en la figura.

Datos: $D = 10{,}0\,m$; $H = 2{,}0\,m$; $|\vec{v}_0| = 15{,}0\,m/s$.

Puede ser útil la siguiente identidad trigonométrica: $1 + \tan^2(x) = \dfrac{1}{\cos^2(x)}$

a) $\theta = 24{,}9^\circ$ &nbsp; b) $\theta = 76{,}4^\circ$ &nbsp; c) $\theta = 21{,}8^\circ$ &nbsp; d) $\theta = 43{,}6^\circ$ &nbsp; e) $\theta = 49{,}8^\circ$

```python
import sympy as sp
from IPython.display import Markdown, display

# Mostrar expresiones sympy como LaTeX dentro de Markdown
latex = lambda e: Markdown(f"$${sp.latex(e)}$$")

# Símbolos y datos
D, H, v0, g = sp.symbols('D H v_0 g', positive=True)
datos = {D: 10, H: 2, v0: 15, g: sp.Rational(98, 10)}

# Incógnitas: componentes de la velocidad inicial y tiempo de llegada a P
v0x, v0y, t, t_P = sp.symbols('v_{0x} v_{0y} t t_P', positive=True)

# Paso 1: Ecuaciones de movimiento (origen en el punto de lanzamiento)
x_t = v0x * t
y_t = v0y * t - sp.Rational(1, 2) * g * t**2
display(Markdown("**Paso 1:** Ecuaciones de movimiento, con origen en el punto de lanzamiento"),
        latex(sp.Eq(sp.Symbol('x(t)'), x_t)), latex(sp.Eq(sp.Symbol('y(t)'), y_t)))

# Paso 2: En t_P el proyectil está en P = (D, H), y |v0| = v_0
eq_x = sp.Eq(x_t.subs(t, t_P), D)
eq_y = sp.Eq(y_t.subs(t, t_P), H)
eq_v = sp.Eq(v0x**2 + v0y**2, v0**2)
display(Markdown("**Paso 2:** Condiciones: en $t_P$ el proyectil está en $P = (D, H)$, y $|\\vec{v}_0| = v_0$"),
        latex(eq_x), latex(eq_y), latex(eq_v))

# Paso 3: Despejar t_P de la ecuación en x
t_P_expr = sp.solve(eq_x, t_P)[0]
display(Markdown("**Paso 3:** Despejar $t_P$ de la ecuación en $x$"),
        latex(sp.Eq(t_P, t_P_expr)))

# Paso 4: Sustituir t_P en la ecuación en y (queda la trayectoria evaluada en P)
eq_y_sub = sp.Eq(sp.expand(eq_y.lhs.subs(t_P, t_P_expr)), H)
display(Markdown("**Paso 4:** Sustituir $t_P$ en la ecuación en $y$"),
        latex(eq_y_sub))

# Paso 5: Sustituir los datos y reducir a una sola ecuación en v0x
display(Markdown("**Paso 5:** Sustituir los datos y reducir a una ecuación con una incógnita"))

# 5a: Valores numéricos de los coeficientes
sistema = [eq_y_sub.subs(datos), eq_v.subs(datos)]
display(Markdown(f"*5a.* Coeficientes con los datos: "
                 f"$D = {datos[D]},\\quad H = {datos[H]},\\quad v_0^2 = {(v0**2).subs(datos)},\\quad "
                 f"\\dfrac{{g D^2}}{{2}} = {(g * D**2 / 2).subs(datos)}$"),
        Markdown("Queda un sistema de 2 ecuaciones con 2 incógnitas:"), *map(latex, sistema))

# 5b: Multiplicar la ecuación en y por v0x² para sacar los denominadores
eq_poly = sp.Eq(sp.expand(sistema[0].lhs * v0x**2), sp.expand(sistema[0].rhs * v0x**2))
display(Markdown("*5b.* Multiplicar la ecuación en $y$ por $v_{0x}^2$ para sacar los denominadores"),
        latex(eq_poly))

# 5c: Despejar v0y (aparece lineal)
v0y_expr = sp.solve(eq_poly, v0y)[0]
display(Markdown("*5c.* Despejar $v_{0y}$, que aparece lineal"),
        latex(sp.Eq(v0y, v0y_expr)))

# 5d: Sustituir v0y en la condición del módulo
eq_mod = sp.Eq(sistema[1].lhs.subs(v0y, v0y_expr), sistema[1].rhs)
display(Markdown("*5d.* Sustituir $v_{0y}$ en $v_{0x}^2 + v_{0y}^2 = v_0^2$"),
        latex(eq_mod))

# 5e: Multiplicar por el denominador y pasar todo a un lado -> bicuadrática en v0x
eq_bicuad = sp.Eq(sp.expand(sp.numer(sp.together(eq_mod.lhs - eq_mod.rhs))), 0)
display(Markdown("*5e.* Multiplicar por el denominador y pasar todo a un lado: queda una **bicuadrática** "
                 "(solo potencias pares de $v_{0x}$)"),
        latex(eq_bicuad))

# 5f: Cambio de variable u = v0x² -> cuadrática
u = sp.symbols('u', positive=True)
eq_u = sp.Eq(eq_bicuad.lhs.subs(v0x**2, u), 0)
display(Markdown("*5f.* Cambio de variable $u = v_{0x}^2$: queda una cuadrática"),
        latex(eq_u))

# Paso 6: Resolver la cuadrática y volver a v0x, v0y -> dos soluciones (tiro bajo y tiro alto)
a, b, c = sp.Poly(eq_u.lhs, u).all_coeffs()
disc = b**2 - 4 * a * c
display(Markdown("**Paso 6:** Resolver la cuadrática con Bhaskara: "
                 "$u = \\dfrac{-b \\pm \\sqrt{\\Delta}}{2a}$, con $\\Delta = b^2 - 4ac$"),
        Markdown(f"$a = {a},\\quad b = {b},\\quad c = {c},\\quad \\Delta = {disc}$"))
sols = []
for u_i in sorted(sp.solve(eq_u, u), reverse=True):
    v0x_i = sp.sqrt(u_i)
    v0y_i = v0y_expr.subs(v0x, v0x_i)
    sols.append({v0x: v0x_i, v0y: v0y_i})
    display(Markdown(f"$u = {u_i.evalf(4)} \\Rightarrow v_{{0x}} = \\sqrt{{u}} = {v0x_i.evalf(3)}"
                     f" \\Rightarrow v_{{0y}} = {v0y_i.evalf(3)}$ (con 5c)"))

# Paso 7: Ángulo y tiempo de llegada de cada solución
display(Markdown("**Paso 7:** Ángulo $\\theta = \\arctan(v_{0y} / v_{0x})$ y tiempo de llegada $t_P = D / v_{0x}$"))
for s in sols:
    theta = sp.deg(sp.atan(s[v0y] / s[v0x])).evalf(3)
    t_P_i = t_P_expr.subs(datos).subs(s).evalf(3)
    display(Markdown(f"$\\theta = {theta}^\\circ,\\quad t_P = {t_P_i}\\,s$"))

# Paso 8: Chequeo del muro: en x = D/2 la altura tiene que ser mayor que H
t_muro = (D / 2) / v0x
display(Markdown("**Paso 8:** Chequeo del muro: cuando $x = D/2$, tiene que ser $y > H$"),
        latex(sp.Eq(sp.Symbol('t_{muro}'), t_muro)))
for s in sols:
    theta = sp.deg(sp.atan(s[v0y] / s[v0x])).evalf(3)
    t_m = t_muro.subs(datos).subs(s).evalf(3)
    y_muro = y_t.subs(datos).subs(s).subs(t, t_m).evalf(3)
    supera = "supera el muro" if y_muro > datos[H] else "choca con el muro"
    display(Markdown(f"$\\theta = {theta}^\\circ:\\ t_{{muro}} = {t_m}\\,s \\Rightarrow y(t_{{muro}}) = {y_muro}\\,m$ → **{supera}**"))

display(Markdown("**Respuesta:** b) $\\theta = 76{,}4^\\circ$"))
```
