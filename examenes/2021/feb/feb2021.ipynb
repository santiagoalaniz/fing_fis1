## Ejericio 1

Se usa dinamita para despedazar una roca apoyada sobre un piso horizontal. Los fragmentos de la roca salen disparados en todas direcciones, y después se encuentran a distancias variables de hasta $50m$ de la explosión (el tamaño de la roca es despreciable en comparación con los $50m$). Determine la velocidad máxima con que salieron disparados los fragmentos. 

```python
import sympy as sp
from IPython.display import display, Markdown, Math

# Datos
X_max = 50  # m (distancia máxima donde del fragmento más lejano)
Y_max = 0   # m (altura máxima del fragmento más lejano)
G = 9.81    # m/s^2 (aceleración de la gravedad)

# Incógnitas
x_max, y_max = sp.symbols('x_max y_max')  # m (distancia y altura máxima)
g = sp.symbols('g')  # m/s^2 (aceleración de la gravedad)
v_x, v_y = sp.symbols('v_x v_y')  # m/s (velocidades)
t_max = sp.symbols('t_max')  # s (tiempo en el que se encuentra el fragmento con la distancia máxima)

# Ecuaciones
eq_v_x_v_y = sp.Eq(v_x, v_y)
eq_v_x = sp.Eq(50, v_x * t_max)
eq_t_max = sp.Eq(0, v_y - g * t_max)
sol_t_max = sp.solve(eq_t_max, t_max)[0]
eq_t_max_solved = sp.Eq(t_max, sol_t_max)

# Sustituciones en ecuaciones
eq_v_x_solved = eq_v_x.subs(v_x, v_y).subs(t_max, sol_t_max)
sol_v_y = sp.solve(eq_v_x_solved, v_y)[1]
eq_v_y_solved = sp.Eq(v_y, sol_v_y).subs(g, G).evalf()

# Mostrar todo en un solo output
display(
  Markdown("### Relaciones de las velocidades cuando la velocidad es máxima:"),
  Math(sp.latex(eq_v_x_v_y)),

  Markdown("### Relación de la velocidad en el eje x con la distancia máxima:"),
  Math(sp.latex(eq_v_x)),

  Markdown("### Tiempo total de vuelo:"),
  Math(sp.latex(eq_t_max)),
  Math(sp.latex(eq_t_max_solved)),

  Markdown("### Velocidad en el eje x:"),
  Math(sp.latex(eq_v_x_solved)),
  Math(sp.latex(eq_v_y_solved)),

  Markdown("**Respuesta: A**")
)
```
