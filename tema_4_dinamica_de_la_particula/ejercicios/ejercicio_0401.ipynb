# Ejercicio 1. Suma de fuerzas. Uso de $g$ y $F_0$ como unidades de aceleración y fuerza.

Sobre un objeto se ejercen las siguientes fuerzas:

$
\vec{F}_1 = \frac{F_0(i-j)}{\sqrt{2}} \quad \text{y} \quad \vec{F}_2 = \frac{3F_0(i+2j)}{\sqrt{2}}
$

Se pide:

- Calcular la aceleracion del objeto si su masa es $m = \frac{F_0}{6g}$.
- Que fuerza $\vec{F}_3$ se debe aplicar al objeto para que su aceleración sea nula.

```python
import sympy as sp

# Coordinates
i, j = sp.symbols('i j')

# Constants
g, f0 = sp.symbols('g F_0')

# force vectors
f_1 = (f0*(i-j))/sp.sqrt(2)
print('Force 1:')
display(f_1)

f_2 = (3*f0*(i+2*j))/sp.sqrt(2)
print('Force 2:')
display(f_2)

# mass
m = f0/(6*g)

# acceleration
a = sp.simplify((f_1 + f_2)/m)
print('Acceleration:')
display(a)

# Force 3 that will make the acceleration zero
f_3 = sp.simplify(-f_1-f_2)
print('Force 3:')
display(f_3)

# acceleration with force 3
a = sp.simplify((f_1 + f_2 + f_3)/m)
print('Acceleration with force 3:')
display(a)
```
