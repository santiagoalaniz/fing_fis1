# Ejercicio 4, Una polea sin masa puede tener aceleracion.

Alguien ejerce una fuerza $F$ directamente hacia arriba sobre el eje de la polea que se muestra en la figura.

![](ejercicio_0404.png)

Considera que la polea no tiene masa y que no hay fricción entre la cuerda y la polea.

Dos objetos $m_1$ y $m_2$ están conectados por una cuerda que pasa por la polea. El objeto $m_1$ tiene una masa de $m_1 = 1.2 \, \text{kg}$ y el objeto $m_2$ tiene una masa de $m_2 = 1.9 \, \text{kg}$.

Observa que si la fuerza $F$ es relativamente pequeña, no levantara a $m_2$ pero si lograria acelerar a $m_1$.

Se pide:

- (a) Cual es el valor mas que grande que puede tener la fuerza $F$ para que $m_2$ no se levante.

- (b) Cual es la tension en la cuerda cuando la fuerza $F$ es igual a $110 \, \text{N}$.

- (c) Con la tension determinada en (b) cual es la aceleración de $m_1$.

# Solucion

## (a) Cual es el valor mas que grande que puede tener la fuerza $F$ para que $m_2$ no se levante.

La fuerza $F$ es una fuerza que se aplica en el eje de la polea que une a $m_1$ con $m_2$. 

El sistema, en su punto critico, justo antes de que $m_2$ se levante, la fuerza normal $N$ que se aplica en $m_2$ sea igual a cero.

$$
\begin{aligned}
T - m_2 g &= 0 \\
T = m_2 g
\end{aligned}
$$

La tension T es una consecuencia de aplicar la fuerza $F$ en el eje de la polea, hay dos fuerzas $T$ de cuerda que se aplican en $m_1$ y $m_2$.

Por lo tanto.

$F = 2 T = 2 m_2 g$

```python
# Constants
m2 = 1.9  # kg
g = 9.8   # m/s^2

# Calculate the maximum force F for which m2 does not lift off
F_max = 2 * m2 * g
F_max

print(f"The maximum force F for which m2 does not lift off is {F_max} N.")
```

## (b) Cual es la tension en la cuerda cuando la fuerza $F$ es igual a $110 \, \text{N}$.

```python
F = 110 # N

# Sabemos que F = 2 T

T = F / 2

print(f"The tension in the string is {T} N.")
```

## (c) Con la tension determinada en (b) cual es la aceleración de $m_1$.

```python
import sympy as sp

# Variables
m1 = 1.2  # kg
T = 55    # N
g = 9.8   # m/s^2

# Symbols
# j coord
j = sp.symbols('j')
# acc vec
a = sp.symbols('a')

x = sp.nsimplify((T - m1 * g)/m1, tolerance=0.01)

a_eq = sp.Eq(a, x*j)

display(a_eq)
```
