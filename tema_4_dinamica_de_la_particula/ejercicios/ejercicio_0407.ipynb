# Problema 7, Poleas moviles y fijas.

Tres objetos se cuelgan de poleas y cuerdas ideales. Calcular la aceleracion de cada objeto y la tension de la cuerda.

```python
import sympy as sp

# Constantes
m1, m2, m3 = 1.0, 2.0, 3.0 # kg
g = sp.Symbol('g') # m/s^2

# Variables
## aceleraciones
a1, a2, a3 = sp.symbols('a1 a2 a3')

## Fuerzas
P1, P2, P3 = -m1*g, -m2*g, -m3*g
T = sp.symbols('T')

# Ecuaciones de movimiento por cada bloque
eq_1 = sp.Eq(m1*a1, P1 + T)
display(eq_1)
eq_2 = sp.Eq(m2*a2, P2 + 2*T)
display(eq_2)
eq_3 = sp.Eq(m3*a3, P3 + T)
display(eq_3)
## Relacion del sistema: inextension de la cuerda 
eq_4 = sp.Eq(a2, -(a3 + a1)/2)

# Resolviendo el sistema de ecuaciones
sol = sp.solve([eq_1, eq_2, eq_3, eq_4], (a1, a2, a3, T))
display(sol)
```
