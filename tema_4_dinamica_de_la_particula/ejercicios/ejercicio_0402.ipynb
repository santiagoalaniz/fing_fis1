# Ejercicio 2

Un camion transporta un pequeño bloque de hielo seco en la caja trasera.

Inicialmente, el camion esta en reposo y acelera. Si observamos que el bloque esta quieto respecto al camion.

- En que parte de la caja trasera del camion se encuentra el bloque de hielo seco?
- Quien ejerce la fuerza que acelera el bloque de hielo seco?

Al cabo de unos segundos el camion y el bloque aceleran a velocidad constante $|v_0| = 5 m/s$

- Que fuerzas actuan sobre el bloque de hielo seco?

Al cabo de unos segundos el camion comienza a frenar a razon de $a_0 = 0.5 m/s^2$.

- Que velocidad tiene el camion segun un peaton inmovil en la acer a los dos segundos?

- Que velocidad tiene el bloque de hielo seco segun el camion a los dos segundos?

#### En que parte de la caja trasera del camion se encuentra el bloque de hielo seco?

El bloque de hielo seco se encuentra en la parte trasera de la caja del camion, por eso es que el camion no percibe que el bloque de hielo seco se mueva.

#### Quien ejerce la fuerza que acelera el bloque de hielo seco?

La fuerza que acelera el bloque de hielo seco es la normal de la puerta de la caja del camion.

#### Que fuerzas actuan sobre el bloque de hielo seco?

Cuando el camion y el bloque de hielo seco aceleran a velocidad constante, no hay aceleracion, es decir, la suma de las fuerzas es igual a cero. Las fuerzas que actuan sobre el bloque de hielo seco son:

- La fuerza normal de la base de la caja del camion
- La fuerza de gravedad

#### Que velocidad tiene el camion segun un peaton inmovil en la acera?

```python
import sympy as sp

# simbolos
i, j = sp.symbols('i j')    # coordenadas unitarias
t = sp.symbols('t')         # tiempo

# constantes 
mod_a0 = 0.5                # m/s^2 (modulo aceleracion inicial)
mod_v0 = 5                  # m/s (modulo velocidad inicial)

# vectores
v_camion_t = mod_v0 - mod_a0*t # el camion va frenando
v_bloque_t = mod_v0

# velocidad a los 2s del camion visto por el peaton
v_camion_2 = sp.nsimplify(v_camion_t.subs(t, 2))
print(f"el camion va a {v_camion_2} m/s a los 2s (visto por el peaton)")

# velocidad a los 2s del bloque de hielo visto por el camion
v_bloque_2 = mod_v0 - v_camion_2
print(f"el bloque de hielo va a {v_bloque_2} m/s a los 2s (visto por el camion)")
```
