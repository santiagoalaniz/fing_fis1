# Ejercicio 5, Un arrastre seguro.

Un automovil de 1200 kg debe ser arrastrado por un plano incliando a 18 grados por medio de un cable atado a la parte trasera de un camion grua.

El cable forma un angulo de 27 grados con el plano inclinado y puede soportar una tension maxima de 4600 N.

Sin que se rompa el cable, cual es la mayor distancia que el automovil puede recorrer los primeros 7.5 segundos despues de arrancar desde el reposo? Desprecie las fuerzas resistivas del automovil.

```python
import numpy as np
import sympy as sp

# Constantes
m = 1200 # kg, masa automovil
mod_t = 4600 # N, modulo de la tension
theta_floor = 18 # grados, angulo del plano inclinado respecto al piso
theta_floor_string = 27 # grados, angulo de la cuerda respecto al plano inclinado

# Centro inercial: eje de la rueda delantera

## analisis de fuerzas eje x
# Tension
mod_t_x = mod_t * np.cos(np.deg2rad(theta_floor_string))
# Peso
mod_p_x = m * 9.81 * np.sin(np.deg2rad(theta_floor))
# Aceleracion
a_x = (mod_t_x - mod_p_x) / m
print(f"Aceleracion en x: {a_x:.2f} m/s^2")

# Constantes
t = sp.Symbol('t') # tiempo

# Solucion
## Funcion de posicion en cada eje
r_i = 1/2*a_x*t**2 # Posicion en x
## Variables
t_sol = 7.5 # s, tiempo de solucion
# Posicion final
r_i_sol = r_i.subs(t, t_sol)
print(f"Desplazamiento final en x: {r_i_sol:.2f} m")
```
