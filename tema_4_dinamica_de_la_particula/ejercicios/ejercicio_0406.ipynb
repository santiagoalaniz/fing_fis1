# Ejercicio 6, Maquina de Atwood.

El sistema de la figura esta formado por una polea de masa despreciable por la que pasa y no desliza una cuerda ideal, que tiene unidas a sus extremos dos masas $m_1$ y $m_2 = 2 m_1$.

La polea puede girar sin friccion alrededor de su centro y la configuracion inicial del problema corresponde a la de la figura, con las masas y la polea en reposo.

![](ejercicio_0406.png)

$L_2 = \frac{3L_1}{4}$

Se libera el sistema. Calcula la velocidad de la masa $m_1$ en el instante que esta a la misma altura que la masa $m_2$.

```python
import sympy as sp

# Constantes
## m1
m_1 = sp.Symbol('m_1') # kg
l_1 = sp.Symbol('l_1') # m
## m2
m_2 = 2*m_1
l_2 = (3/4) * l_1
## universales
g = sp.Symbol('g') # m/s^2

# Variables
t = sp.Symbol('t') # tiempo
j = sp.symbols('j') # unitario

# Analisis de fuerzas.
## m1
P_1 = - m_1 * g 
T_1 = m_2 * g 
## m2
P_2 = - m_2 * g 
T_2 = m_1 * g 

# Ecuaciones de movimiento

## aceleracion
### m1
a_1 = (P_1 + T_1) / m_1 # m/s^2
display(sp.Eq(sp.symbols('a_1'), a_1))
### m2
a_2 = (P_2 + T_2) / m_2 # m/s^2
display(sp.Eq(sp.symbols('a_2'), a_2))

## velocidad
### Condiciones iniciales
v0_1, v0_2 = 0, 0
### m1
v_1 = v0_1 + a_1 * t
display(sp.Eq(sp.symbols('v_1'), v_1))
### m2
v_2 = v0_2 + a_2 * t
display(sp.Eq(sp.symbols('v_2'), v_2))

## posicion
### Condiciones iniciales
x0_1, x0_2 = -l_1, -l_2
### m1
x_1 = x0_1  + (1/2) * a_1 * t**2
display(sp.Eq(sp.symbols('x_1'), x_1))
### m2
x_2 = x0_2 + (1/2) * a_2 * t**2
display(sp.Eq(sp.symbols('x_2'), x_2))

# Solucion: calcular v_1, v_2 (t_sol), donde x_1(t_sol) = x_2(t_sol), t_sol > 0
t_sol = sp.simplify(sp.solve(x_1 - x_2, t)[1])
display(sp.Eq(sp.symbols('t_sol'), t_sol))
v1_sol = v_1.subs(t, t_sol)
display(sp.Eq(sp.symbols('v1_sol'), v1_sol))
v2_sol = v_2.subs(t, t_sol)
display(sp.Eq(sp.symbols('v2_sol'), v2_sol))


```
