# Ejercicio 3. Frecuencia y velocidad tangencial. Trineo.

Un trineo recorre una pista circular de 10m de radio, se desplaza a una velocidad de 10rev/s. Repentinamente se detiene en un punto de su recorrido.

Un paquete, que estaba en el trineo, continua moviendose a la velocidad que tenia justo antes de parar.

Calcular:

- Cual es la posicion del paquete luego de 2s?

```python
import sympy as sp
import numpy as np

# Constantes
RADIO_PISTA = 10 # metros
REV_CARRO = 0.10 # rev/s

# Calculos de velocidad
vel_angular = 2 * np.pi * REV_CARRO
vel_tangencial = RADIO_PISTA * vel_angular # 2pi m/s

# Eje de coordenadas, centro de la pista.
# Hipotesis: 
# - Movimiento anti-horario
# - El carro para en el punto (10, 0)
# - El vector de velocidad es tangente a la circunferencia que pasa por el punto (10, 0)
# - Por lo tanto, el vector de velocidad es paralelo al eje y, sentido positivo

# Tiempo
t = sp.symbols('t')

# Velocidad inicial del paquete
v_x = 0
v_y = vel_tangencial # m/s (constante)

# Posicion inicial del paquete
r0_x = 10
r0_y = 0

# Funciones de posicion del paquete respecto al tiempo
r_x_t = r0_x + v_x * t # esto es constante pq no hay aceleracion en x
r_y_t = r0_y + v_y * t # esto es line

# Posicion del paquete despues de 2 segundos
r_x_2 = r_x_t.subs(t, 2)
r_y_2 = r_y_t.subs(t, 2)

# Distancia al centro de la pista
distancia = sp.sqrt(r_x_2**2 + r_y_2**2)

# Angulo recta que pasa por el paquete y el centro de la pista
angulo = sp.atan(r_y_2 / r_x_2)
angulo_grados = angulo * 180 / np.pi

print(f"Posicion del paquete despues de 2 segundos: (%.2f, %.2f)" % (r_x_2, r_y_2))
print(f"Distancia al centro de la pista: {distancia:.2f} metros")
print(f"Angulo de la recta que pasa por el paquete y el centro de la pista: {angulo:.2f} radianes")
print(f"Angulo de la recta que pasa por el paquete y el centro de la pista: {angulo_grados:.2f} grados")
```
